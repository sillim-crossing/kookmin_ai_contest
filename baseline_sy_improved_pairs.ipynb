```python
from scipy.stats import pearsonr

def safe_corr_with_pvalue(x, y):
    if np.std(x) == 0 or np.std(y) == 0:
        return 0.0, 1.0
    corr, p_value = pearsonr(x, y)
    return float(corr), float(p_value) # 피어슨 상관계수, p-value
```

```python
def calculate_score_advanced(corr, p_value, lag):
    """
    상관계수(70%) + p-value(20%) + lag(10%)
    """
    # 상관계수 (0~70점)
    corr_score = abs(corr) * 70

    # 통계적 유의성 (0~20점)
    if p_value < 0.001:
        sig_score = 20
    elif p_value < 0.01:
        sig_score = 15
    elif p_value < 0.05:
        sig_score = 10
    else:
        sig_score = 5

    # Lag (0~10점) - 짧을수록 좋음
    if lag <= 2:
        lag_score = 10
    elif lag <= 4:
        lag_score = 7
    elif lag <= 6:
        lag_score = 4
    else:
        lag_score = 1

    return corr_score + sig_score + lag_score
```

## DTW 값이 작을수록 패턴이 유사

DTW (Dynamic Time Warping)은 시계열 데이터 간의 유사성을 비교하기 위한 알고리즘이다. DTW는 시계열 데이터 간의 길이나 속도가 달라도, 이것을 고려하여 유사성을 측정할 수 있기 때문에 시계열 데이터 분석에 많이 활용된다.

왜 DTW를 쓰는가?
상관계수만으로는 놓치는 것:

- 속도 차이 (A는 천천히, B는 빠르게 변함)
- 타이밍 차이 (A가 먼저 변하고 B가 조금 늦게 따라감)
- 노이즈 (한두 달 튀는 값)

DTW는 이런 것까지 고려해서 "진짜 패턴 유사도"를 측정

# 이게 새로운 코드입니다! 상관계수 기반 구조를 더 잘 유지한다고 해요..

```python
from tqdm import tqdm
import numpy as np
from sklearn.preprocessing import StandardScaler


# -------------------------------------------------
# DTW 함수
# -------------------------------------------------
def dtw_distance(series_A, series_B):
    n, m = len(series_A), len(series_B)
    dtw_matrix = np.full((n + 1, m + 1), np.inf)
    dtw_matrix[0, 0] = 0

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = abs(series_A[i-1] - series_B[j-1])
            dtw_matrix[i, j] = cost + min(
                dtw_matrix[i-1, j],
                dtw_matrix[i, j-1],
                dtw_matrix[i-1, j-1]
            )
    return dtw_matrix[n, m]


def calculate_normalized_dtw(x, y, lag):
    if len(x) <= lag or len(y) <= lag:
        return float('inf')

    x_aligned = x[:-lag]
    y_aligned = y[lag:]

    if len(x_aligned) < 4:
        return float('inf')

    scaler = StandardScaler()
    nx = scaler.fit_transform(x_aligned.reshape(-1, 1)).flatten()
    ny = scaler.fit_transform(y_aligned.reshape(-1, 1)).flatten()

    d = dtw_distance(nx, ny)
    return d / len(nx)


# =====================================================================
# ⭐ DTW 안전 버전 + 초저품질 pair 제거(3개 규칙)
# =====================================================================
def find_comovement_pairs_safe_dtw(
    pivot,
    max_lag=9,
    min_nonzero=12,
    corr_threshold=0.35,
    score_threshold=35,
    use_dtw=True
):

    items = pivot.index.to_list()
    months = pivot.columns.to_list()
    n_months = len(months)

    results = []

    for i, leader in tqdm(enumerate(items), total=len(items), desc="Finding pairs"):
        x = pivot.loc[leader].values.astype(float)
        if np.count_nonzero(x) < min_nonzero:
            continue

        for follower in items:
            if follower == leader:
                continue

            y = pivot.loc[follower].values.astype(float)
            if np.count_nonzero(y) < min_nonzero:
                continue

            # -------------------------------------------------
            # ⭐ (A) 초저품질 필터 1: zero 비율이 70% 이상
            # -------------------------------------------------
            zero_x = (x == 0).sum() / len(x)
            zero_y = (y == 0).sum() / len(y)
            if max(zero_x, zero_y) >= 0.70:
                continue

            # -------------------------------------------------
            # ⭐ (B) 초저품질 필터 2: CV 비정상적으로 큰 경우
            # -------------------------------------------------
            cv_x = np.std(x) / (np.mean(x) + 1)
            cv_y = np.std(y) / (np.mean(y) + 1)
            if max(cv_x, cv_y) >= 5:
                continue

            # -------------------------------------------------
            # ⭐ (C) 초저품질 필터 3: 초기 lag 안정성
            # -------------------------------------------------
            lag_corrs = []
            for lag in range(1, min(4, max_lag) + 1):
                if len(x) > lag and len(y) > lag:
                    c, _ = safe_corr_with_pvalue(x[:-lag], y[lag:])
                    if not np.isnan(c):
                        lag_corrs.append(c)

            if len(lag_corrs) >= 2:
                if np.max(lag_corrs) - np.min(lag_corrs) > 0.85:
                    continue

            # -------------------------------------------------
            # 1) 최고 상관계수 찾기
            # -------------------------------------------------
            best_corr = 0.0
            best_lag = None
            best_p_value = 1.0

            for lag in range(1, max_lag + 1):
                if len(x) <= lag or len(y) <= lag:
                    continue

                corr, p_value = safe_corr_with_pvalue(x[:-lag], y[lag:])
                if abs(corr) > abs(best_corr):
                    best_corr = corr
                    best_lag = lag
                    best_p_value = p_value

            # 2) 상관계수 1차 필터
            if best_lag is None or abs(best_corr) < corr_threshold:
                continue

            # 3) corr + p-value + lag 기반 점수
            score = calculate_score_advanced(best_corr, best_p_value, best_lag)

            # -------------------------------------------------
            # ⭐ 4) DTW는 절대필터 ❌ / 보너스만 ⭕
            # -------------------------------------------------
            if use_dtw:
                dtw_dist = calculate_normalized_dtw(x, y, best_lag)

                if dtw_dist == float('inf'):
                    dtw_bonus = 1.0
                else:
                    dtw_bonus = 1 / (1 + dtw_dist * 0.5)

                score = score * (1 + dtw_bonus)
            else:
                dtw_dist = None

            # 5) 최종 점수 필터
            if score >= score_threshold:
                results.append({
                    "leading_item_id": leader,
                    "following_item_id": follower,
                    "best_lag": best_lag,
                    "max_corr": best_corr,
                    "p_value": best_p_value,
                    "dtw_distance": dtw_dist,
                    "comovement_score": score
                })

    return pd.DataFrame(results)
```

```python
pairs_with_dtw = find_comovement_pairs_safe_dtw(
    pivot=pivot,
    max_lag=9,
    min_nonzero=12,
    corr_threshold=0.33,
    score_threshold=40,
    use_dtw=False    # DTW 보너스 적용 (근데 하나 안하나 차이 있는지는 확인 필요)
)

print(len(pairs_with_dtw))
print(pairs_with_dtw.head())
```

```python
# ============================================================
#  ⬇⬇⬇⬇⬇⬇⬇⬇  여기에 넣어라: Pairs 생성 직후 위치  ⬇⬇⬇⬇⬇⬇⬇⬇
# ============================================================

# 1) pivot 기반 전체 item graph 생성
import networkx as nx
import numpy as np
from community import community_louvain  # Louvain

# --- 그래프 생성 ---
def build_item_graph(pivot_value, corr_thr=0.3, lag_thr=0.2, max_lag=3):

    items = pivot_value.columns.tolist()
    G = nx.DiGraph()

    # Pearson corr
    corr_matrix = pivot_value.corr()

    # lagged corr
    def lag_corr(a, b):
        a = a - a.mean()
        b = b - b.mean()
        best = 0
        for lag in range(1, max_lag + 1):
            if len(a) <= lag: break
            c = np.corrcoef(a[:-lag], b[lag:])[0, 1]
            best = max(best, abs(c))
        return best

    for i in items:
        for j in items:
            if i == j: continue
            w1 = corr_matrix.loc[i, j]
            if abs(w1) >= corr_thr:
                G.add_edge(i, j, weight=float(w1))

            w2 = lag_corr(pivot_value[i].values, pivot_value[j].values)
            if w2 >= lag_thr:
                G.add_edge(i, j, weight=float(w2))

    return G


# pivot_value는 year_month × item_id 형태이므로 다시 transpose
pivot_value_for_graph = pivot_value.copy()

G = build_item_graph(pivot_value_for_graph)

print("그래프 노드수:", G.number_of_nodes())
print("그래프 엣지수:", G.number_of_edges())


# --- Louvain Community ---
partition = community_louvain.best_partition(G.to_undirected(), weight='weight')
louvain_df = pd.DataFrame({'item_id': list(partition.keys()),
                           'community': list(partition.values())})


# --- Greedy Modularity ---
greedy_com = nx.algorithms.community.greedy_modularity_communities(G.to_undirected())
greedy_assign = {}
for cid, group in enumerate(greedy_com):
    for node in group:
        greedy_assign[node] = cid

greedy_df = pd.DataFrame({'item_id': list(greedy_assign.keys()),
                          'community': list(greedy_assign.values())})


# --- Centrality ---
in_deg  = dict(G.in_degree(weight='weight'))
out_deg = dict(G.out_degree(weight='weight'))
close   = nx.closeness_centrality(G)
btwn    = nx.betweenness_centrality(G, weight='weight')

centrality_df = pd.DataFrame({
    'item_id': list(G.nodes()),
    'in_degree': [in_deg.get(i, 0) for i in G.nodes()],
    'out_degree': [out_deg.get(i, 0) for i in G.nodes()],
    'closeness': [close.get(i, 0) for i in G.nodes()],
    'betweenness': [btwn.get(i, 0) for i in G.nodes()]
})


# -------------------------
# 2) pair_df에 feature merge
# -------------------------

# dict 형태로 변환
louvain_map = dict(zip(louvain_df['item_id'], louvain_df['community']))
greedy_map  = dict(zip(greedy_df['item_id'], greedy_df['community']))
cent_map    = centrality_df.set_index('item_id').to_dict(orient='index')


def add_graph_features(row):

    A = row.leading_item_id
    B = row.following_item_id

    # Louvain
    cA = louvain_map.get(A, -1)
    cB = louvain_map.get(B, -1)

    # Greedy
    gA = greedy_map.get(A, -1)
    gB = greedy_map.get(B, -1)

    # centrality
    CA = cent_map.get(A, {'in_degree':0,'out_degree':0,'closeness':0,'betweenness':0})
    CB = cent_map.get(B, {'in_degree':0,'out_degree':0,'closeness':0,'betweenness':0})

    return pd.Series({
        'lead_comm': cA,
        'foll_comm': cB,
        'same_comm': int(cA == cB),

        'lead_greedy': gA,
        'foll_greedy': gB,
        'same_greedy': int(gA == gB),

        'lead_in_deg': CA['in_degree'],
        'foll_in_deg': CB['in_degree'],
        'deg_gap': CA['in_degree'] - CB['in_degree'],
        'deg_ratio': CA['in_degree'] / (CB['in_degree'] + 1e-6),

        'lead_btwn': CA['betweenness'],
        'foll_btwn': CB['betweenness'],
        'btwn_gap': CA['betweenness'] - CB['betweenness'],
        'btwn_ratio': CA['betweenness'] / (CB['betweenness'] + 1e-6),
    })


graph_feats = pairs_with_dtw.apply(add_graph_features, axis=1)
pairs_with_dtw = pd.concat([pairs_with_dtw, graph_feats], axis=1)

print("그래프 기반 feature 추가 완료!")
print(pairs_with_dtw.head())
```

# 기존 추출된 pair들, 네트워크 극단값 기반으로 저품질 제거하고 모델링에 넣기 (옵션2 - 데이터 품질 향상)

```python
### 안쓸땐 주석처리

# === (A) 그래프 기반 재가중치 + pair 필터링 (Stage1 품질 개선) ===
import numpy as np

# 그래프 피처까지 merge가 끝난 최종 pair 테이블
# columns 예시: leading_item_id, following_item_id, best_lag, max_corr,
#               dtw_distance, comovement_score,
#               lead_comm, foll_comm, same_comm,
#               lead_greedy, foll_greedy, same_greedy,
#               lead_in_deg, foll_in_deg, deg_gap, deg_ratio,
#               lead_btwn, foll_btwn, btwn_gap, btwn_ratio
pairs_for_model = pairs_with_dtw.copy()

# 1) deg_ratio, btwn_ratio 이상치/극단값 클리핑
deg_ratio_clipped = pairs_for_model['deg_ratio'].clip(0.3, 3.0)
btwn_ratio_raw = pairs_for_model['btwn_ratio'].replace([np.inf, -np.inf], np.nan).fillna(1.0)
btwn_ratio_clipped = btwn_ratio_raw.clip(0.3, 3.0)

# 2) 그래프 기반 가중 score 정의
#    - 기본: comovement_score
#    - same_comm == 1 인 쌍에 약간 보너스
#    - deg_ratio, btwn_ratio는 너무 세게 쓰지 않고 약한 지수로 반영
pairs_for_model['graph_score'] = (
    pairs_for_model['comovement_score']
    * (1.05 + 0.05 * pairs_for_model['same_comm'])     # same_comm 1이면 대략 10% 보너스
    * (deg_ratio_clipped / 1.5) ** 0.3                 # degree 비례 약하게 반영
    * (btwn_ratio_clipped) ** 0.2                      # betweenness 비례 아주 약하게 반영
)

# 3) leader별 상위 TOP_K 개 pair만 사용 (노이즈 쌍 컷)
TOP_K = 35  # 필요하면 바꿔보면서 실험

pairs_for_model = (
    pairs_for_model
    .sort_values(['leading_item_id', 'graph_score'], ascending=[True, False])
    .groupby('leading_item_id')
    .head(TOP_K)
    .reset_index(drop=True)
)

print("그래프 노드수는 그대로, 재가중치 후 사용될 pair 수:", len(pairs_for_model))
pairs_for_model.head()
```

# 이거는 그냥 제가 살펴보려고 한 코드니까 실행안하셔도 됩니다!

```python
"""
PAIR 품질 분포 상세 분석
- 상관계수 분포
- DTW 거리 분포
- 품질 점수 분포
- 저품질 vs 고품질 특성
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 한글 폰트 설정 (선택사항)
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False

print("="*80)
print("⭐ PAIR 품질 분포 분석")
print("="*80)

# ============================================================================
# 1. 기본 통계
# ============================================================================

print("\n" + "="*80)
print("1. 기본 통계")
print("="*80)

print(f"\n전체 PAIR 개수: {len(pairs_with_dtw):,}개")
print(f"\n[상관계수 통계]")
print(f"  평균: {pairs_with_dtw['max_corr'].mean():.4f}")
print(f"  중앙값: {pairs_with_dtw['max_corr'].median():.4f}")
print(f"  표준편차: {pairs_with_dtw['max_corr'].std():.4f}")
print(f"  최소: {pairs_with_dtw['max_corr'].min():.4f}")
print(f"  최대: {pairs_with_dtw['max_corr'].max():.4f}")
print(f"  Q1 (25%): {pairs_with_dtw['max_corr'].quantile(0.25):.4f}")
print(f"  Q3 (75%): {pairs_with_dtw['max_corr'].quantile(0.75):.4f}")

print(f"\n[DTW 거리 통계]")
print(f"  평균: {pairs_with_dtw['dtw_distance'].mean():.2f}")
print(f"  중앙값: {pairs_with_dtw['dtw_distance'].median():.2f}")
print(f"  표준편차: {pairs_with_dtw['dtw_distance'].std():.2f}")
print(f"  최소: {pairs_with_dtw['dtw_distance'].min():.2f}")
print(f"  최대: {pairs_with_dtw['dtw_distance'].max():.2f}")
print(f"  Q1 (25%): {pairs_with_dtw['dtw_distance'].quantile(0.25):.2f}")
print(f"  Q3 (75%): {pairs_with_dtw['dtw_distance'].quantile(0.75):.2f}")

# ============================================================================
# 2. 품질 등급 분류
# ============================================================================

print("\n" + "="*80)
print("2. 품질 등급 분류")
print("="*80)

# 상관계수 기반 등급
pairs_analysis = pairs_with_dtw.copy()

# 4단계 등급
pairs_analysis['quality_grade'] = pd.cut(
    pairs_analysis['max_corr'],
    bins=[0, 0.5, 0.7, 0.85, 1.0],
    labels=['저품질 (0-0.5)', '중품질 (0.5-0.7)', '고품질 (0.7-0.85)', '최고품질 (0.85+)']
)

print("\n[등급별 PAIR 개수 및 비율]")
grade_counts = pairs_analysis['quality_grade'].value_counts().sort_index()
for grade, count in grade_counts.items():
    pct = count / len(pairs_analysis) * 100
    print(f"  {grade}: {count:,}개 ({pct:.1f}%)")

# 등급별 DTW 통계
print("\n[등급별 평균 DTW 거리]")
dtw_by_grade = pairs_analysis.groupby('quality_grade')['dtw_distance'].agg(['mean', 'std', 'median'])
print(dtw_by_grade.round(2))

# 등급별 Lag 통계
print("\n[등급별 평균 Lag]")
lag_by_grade = pairs_analysis.groupby('quality_grade')['best_lag'].agg(['mean', 'median'])
print(lag_by_grade.round(2))

# ============================================================================
# 3. 상위/하위 PAIR 분석
# ============================================================================

print("\n" + "="*80)
print("3. 상위/하위 PAIR 분석")
print("="*80)

# 상위 10개
print("\n[Top 10 고품질 PAIR (상관계수 높은 순)]")
top10 = pairs_analysis.nlargest(10, 'max_corr')[['leading_item_id', 'following_item_id', 'max_corr', 'best_lag', 'dtw_distance']]
for i, row in enumerate(top10.itertuples(), 1):
    print(f"  {i:2d}. Leader {row.leading_item_id} → Follower {row.following_item_id}")
    print(f"      Corr: {row.max_corr:.4f}, Lag: {row.best_lag}, DTW: {row.dtw_distance:.1f}")

# 하위 10개
print("\n[Bottom 10 저품질 PAIR (상관계수 낮은 순)]")
bottom10 = pairs_analysis.nsmallest(10, 'max_corr')[['leading_item_id', 'following_item_id', 'max_corr', 'best_lag', 'dtw_distance']]
for i, row in enumerate(bottom10.itertuples(), 1):
    print(f"  {i:2d}. Leader {row.leading_item_id} → Follower {row.following_item_id}")
    print(f"      Corr: {row.max_corr:.4f}, Lag: {row.best_lag}, DTW: {row.dtw_distance:.1f}")

# ============================================================================
# 4. 필터링 시뮬레이션
# ============================================================================

print("\n" + "="*80)
print("4. 필터링 시뮬레이션 (다양한 임계값)")
print("="*80)

thresholds = [0.5, 0.6, 0.7, 0.8, 0.85, 0.9]

print("\n[상관계수 임계값별 PAIR 유지율]")
print(f"{'임계값':>8s} {'유지 개수':>10s} {'유지율':>8s} {'평균 DTW':>10s}")
print("-" * 40)
for thresh in thresholds:
    filtered = pairs_analysis[pairs_analysis['max_corr'] >= thresh]
    count = len(filtered)
    pct = count / len(pairs_analysis) * 100
    avg_dtw = filtered['dtw_distance'].mean()
    print(f"{thresh:>8.2f} {count:>10,d} {pct:>7.1f}% {avg_dtw:>10.1f}")

# 현재 2,700개 목표에 맞는 임계값
target_count = 2700
sorted_corr = pairs_analysis['max_corr'].sort_values(ascending=False)
if len(sorted_corr) >= target_count:
    optimal_threshold = sorted_corr.iloc[target_count]
    print(f"\n⭐ 2,700개 유지 최적 임계값: {optimal_threshold:.4f}")
else:
    print(f"\n⚠️  전체 PAIR가 {len(pairs_analysis)}개로 2,700개 미만")

# ============================================================================
# 5. 상관계수 vs DTW 관계 분석
# ============================================================================

print("\n" + "="*80)
print("5. 상관계수 vs DTW 관계 분석")
print("="*80)

# 상관관계
corr_dtw_relationship = pairs_analysis[['max_corr', 'dtw_distance']].corr().iloc[0, 1]
print(f"\n상관계수 vs DTW 거리 상관관계: {corr_dtw_relationship:.4f}")

if corr_dtw_relationship < -0.5:
    print("  → 강한 음의 상관 (높은 상관 = 낮은 DTW) ✅")
elif corr_dtw_relationship < -0.3:
    print("  → 중간 음의 상관")
else:
    print("  → 약한 상관 (독립적인 메트릭)")

# 사분면 분석
high_corr_threshold = pairs_analysis['max_corr'].quantile(0.75)
low_dtw_threshold = pairs_analysis['dtw_distance'].quantile(0.25)

print(f"\n[사분면 분석]")
print(f"  고상관 기준: {high_corr_threshold:.4f}")
print(f"  저DTW 기준: {low_dtw_threshold:.1f}")

q1 = len(pairs_analysis[(pairs_analysis['max_corr'] >= high_corr_threshold) &
                        (pairs_analysis['dtw_distance'] <= low_dtw_threshold)])
q2 = len(pairs_analysis[(pairs_analysis['max_corr'] >= high_corr_threshold) &
                        (pairs_analysis['dtw_distance'] > low_dtw_threshold)])
q3 = len(pairs_analysis[(pairs_analysis['max_corr'] < high_corr_threshold) &
                        (pairs_analysis['dtw_distance'] <= low_dtw_threshold)])
q4 = len(pairs_analysis[(pairs_analysis['max_corr'] < high_corr_threshold) &
                        (pairs_analysis['dtw_distance'] > low_dtw_threshold)])

print(f"\n  Q1 (고상관 + 저DTW): {q1:,}개 ({q1/len(pairs_analysis)*100:.1f}%) ⭐ 최고품질")
print(f"  Q2 (고상관 + 고DTW): {q2:,}개 ({q2/len(pairs_analysis)*100:.1f}%)")
print(f"  Q3 (저상관 + 저DTW): {q3:,}개 ({q3/len(pairs_analysis)*100:.1f}%)")
print(f"  Q4 (저상관 + 고DTW): {q4:,}개 ({q4/len(pairs_analysis)*100:.1f}%) ⚠️ 저품질")

# ============================================================================
# 6. 시각화
# ============================================================================

print("\n" + "="*80)
print("6. 시각화 생성 중...")
print("="*80)

fig = plt.figure(figsize=(18, 12))

# 1. 상관계수 히스토그램
ax1 = plt.subplot(3, 3, 1)
pairs_analysis['max_corr'].hist(bins=50, edgecolor='black', alpha=0.7)
ax1.axvline(pairs_analysis['max_corr'].median(),
            color='red', linestyle='--', linewidth=2,
            label=f'Median: {pairs_analysis["max_corr"].median():.3f}')
ax1.axvline(pairs_analysis['max_corr'].quantile(0.25),
            color='orange', linestyle='--', linewidth=2,
            label=f'Q1: {pairs_analysis["max_corr"].quantile(0.25):.3f}')
ax1.axvline(pairs_analysis['max_corr'].quantile(0.75),
            color='green', linestyle='--', linewidth=2,
            label=f'Q3: {pairs_analysis["max_corr"].quantile(0.75):.3f}')
ax1.set_title('Correlation Distribution', fontsize=12, fontweight='bold')
ax1.set_xlabel('Max Correlation')
ax1.set_ylabel('Count')
ax1.legend(fontsize=8)
ax1.grid(True, alpha=0.3)

# 2. DTW 히스토그램
ax2 = plt.subplot(3, 3, 2)
pairs_analysis['dtw_distance'].hist(bins=50, edgecolor='black', alpha=0.7, color='orange')
ax2.axvline(pairs_analysis['dtw_distance'].median(),
            color='red', linestyle='--', linewidth=2,
            label=f'Median: {pairs_analysis["dtw_distance"].median():.1f}')
ax2.set_title('DTW Distance Distribution', fontsize=12, fontweight='bold')
ax2.set_xlabel('DTW Distance')
ax2.set_ylabel('Count')
ax2.legend(fontsize=8)
ax2.grid(True, alpha=0.3)

# 3. 등급별 파이차트
ax3 = plt.subplot(3, 3, 3)
grade_counts.plot(kind='pie', autopct='%1.1f%%', ax=ax3,
                  colors=['#ff9999', '#ffcc99', '#99ccff', '#99ff99'])
ax3.set_title('Quality Grade Distribution', fontsize=12, fontweight='bold')
ax3.set_ylabel('')

# 4. 상관계수 박스플롯 (등급별)
ax4 = plt.subplot(3, 3, 4)
pairs_analysis.boxplot(column='max_corr', by='quality_grade', ax=ax4)
ax4.set_title('Correlation by Quality Grade', fontsize=12, fontweight='bold')
ax4.set_xlabel('Quality Grade')
ax4.set_ylabel('Max Correlation')
plt.suptitle('')
ax4.tick_params(axis='x', rotation=45)

# 5. DTW 박스플롯 (등급별)
ax5 = plt.subplot(3, 3, 5)
pairs_analysis.boxplot(column='dtw_distance', by='quality_grade', ax=ax5)
ax5.set_title('DTW Distance by Quality Grade', fontsize=12, fontweight='bold')
ax5.set_xlabel('Quality Grade')
ax5.set_ylabel('DTW Distance')
plt.suptitle('')
ax5.tick_params(axis='x', rotation=45)

# 6. Correlation vs DTW 산점도
ax6 = plt.subplot(3, 3, 6)
scatter = ax6.scatter(pairs_analysis['max_corr'],
                     pairs_analysis['dtw_distance'],
                     c=pairs_analysis['best_lag'],
                     alpha=0.3, s=10, cmap='viridis')
ax6.set_xlabel('Max Correlation')
ax6.set_ylabel('DTW Distance')
ax6.set_title('Correlation vs DTW (colored by Lag)', fontsize=12, fontweight='bold')
ax6.grid(True, alpha=0.3)
cbar = plt.colorbar(scatter, ax=ax6)
cbar.set_label('Lag', rotation=270, labelpad=15)

# 7. 등급별 개수 막대그래프
ax7 = plt.subplot(3, 3, 7)
grade_counts.plot(kind='bar', ax=ax7, color=['#ff9999', '#ffcc99', '#99ccff', '#99ff99'])
ax7.set_title('PAIR Count by Quality Grade', fontsize=12, fontweight='bold')
ax7.set_xlabel('Quality Grade')
ax7.set_ylabel('Count')
ax7.tick_params(axis='x', rotation=45)
ax7.grid(True, alpha=0.3, axis='y')

# 값 표시
for i, v in enumerate(grade_counts.values):
    ax7.text(i, v + 20, f'{v:,}', ha='center', va='bottom', fontsize=9)

# 8. 필터링 시뮬레이션 그래프
ax8 = plt.subplot(3, 3, 8)
filter_data = []
for thresh in np.arange(0.3, 1.0, 0.05):
    count = len(pairs_analysis[pairs_analysis['max_corr'] >= thresh])
    filter_data.append({'threshold': thresh, 'count': count})
filter_df = pd.DataFrame(filter_data)

ax8.plot(filter_df['threshold'], filter_df['count'], marker='o', linewidth=2)
ax8.axhline(2700, color='red', linestyle='--', linewidth=2, label='Target: 2,700')
ax8.set_xlabel('Correlation Threshold')
ax8.set_ylabel('Remaining PAIRs')
ax8.set_title('Filtering Simulation', fontsize=12, fontweight='bold')
ax8.legend()
ax8.grid(True, alpha=0.3)

# 9. 사분면 분석 시각화
ax9 = plt.subplot(3, 3, 9)
colors = []
for _, row in pairs_analysis.iterrows():
    if row['max_corr'] >= high_corr_threshold and row['dtw_distance'] <= low_dtw_threshold:
        colors.append('#00ff00')  # Q1: 녹색 (최고)
    elif row['max_corr'] >= high_corr_threshold and row['dtw_distance'] > low_dtw_threshold:
        colors.append('#ffff00')  # Q2: 노란색
    elif row['max_corr'] < high_corr_threshold and row['dtw_distance'] <= low_dtw_threshold:
        colors.append('#ff9900')  # Q3: 주황색
    else:
        colors.append('#ff0000')  # Q4: 빨간색 (최저)

ax9.scatter(pairs_analysis['max_corr'], pairs_analysis['dtw_distance'],
           c=colors, alpha=0.4, s=10)
ax9.axvline(high_corr_threshold, color='black', linestyle='--', linewidth=1)
ax9.axhline(low_dtw_threshold, color='black', linestyle='--', linewidth=1)
ax9.set_xlabel('Max Correlation')
ax9.set_ylabel('DTW Distance')
ax9.set_title('Quadrant Analysis', fontsize=12, fontweight='bold')
ax9.text(0.85, pairs_analysis['dtw_distance'].max() * 0.9, f'Q1: {q1}',
        bbox=dict(boxstyle='round', facecolor='green', alpha=0.5), fontsize=9)
ax9.text(0.85, pairs_analysis['dtw_distance'].max() * 0.5, f'Q2: {q2}',
        bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.5), fontsize=9)
ax9.text(0.3, pairs_analysis['dtw_distance'].max() * 0.9, f'Q3: {q3}',
        bbox=dict(boxstyle='round', facecolor='orange', alpha=0.5), fontsize=9)
ax9.text(0.3, pairs_analysis['dtw_distance'].max() * 0.5, f'Q4: {q4}',
        bbox=dict(boxstyle='round', facecolor='red', alpha=0.5), fontsize=9)
ax9.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('pair_quality_analysis.png', dpi=150, bbox_inches='tight')
print("✓ 시각화 저장: pair_quality_analysis.png")

# ============================================================================
# 7. 상세 리포트 저장
# ============================================================================

print("\n" + "="*80)
print("7. 상세 리포트 저장 중...")
print("="*80)

# CSV 저장
pairs_analysis_export = pairs_analysis.copy()
pairs_analysis_export = pairs_analysis_export.sort_values('max_corr', ascending=False)
pairs_analysis_export.to_csv('pair_quality_detailed.csv', index=False)
print("✓ 상세 데이터 저장: pair_quality_detailed.csv")

# 요약 통계 저장
summary_stats = pd.DataFrame({
    '지표': ['개수', '평균', '중앙값', '표준편차', '최소', '최대', 'Q1', 'Q3'],
    '상관계수': [
        len(pairs_analysis),
        pairs_analysis['max_corr'].mean(),
        pairs_analysis['max_corr'].median(),
        pairs_analysis['max_corr'].std(),
        pairs_analysis['max_corr'].min(),
        pairs_analysis['max_corr'].max(),
        pairs_analysis['max_corr'].quantile(0.25),
        pairs_analysis['max_corr'].quantile(0.75)
    ],
    'DTW거리': [
        len(pairs_analysis),
        pairs_analysis['dtw_distance'].mean(),
        pairs_analysis['dtw_distance'].median(),
        pairs_analysis['dtw_distance'].std(),
        pairs_analysis['dtw_distance'].min(),
        pairs_analysis['dtw_distance'].max(),
        pairs_analysis['dtw_distance'].quantile(0.25),
        pairs_analysis['dtw_distance'].quantile(0.75)
    ]
})
summary_stats.to_csv('pair_quality_summary.csv', index=False)
print("✓ 요약 통계 저장: pair_quality_summary.csv")

# ============================================================================
# 8. 최종 결론 및 권장사항
# ============================================================================

print("\n" + "="*80)
print("⭐ 최종 결론 및 권장사항")
print("="*80)

high_quality_ratio = (pairs_analysis['max_corr'] > 0.7).sum() / len(pairs_analysis) * 100
medium_quality_ratio = ((pairs_analysis['max_corr'] > 0.5) & (pairs_analysis['max_corr'] <= 0.7)).sum() / len(pairs_analysis) * 100
low_quality_ratio = (pairs_analysis['max_corr'] <= 0.5).sum() / len(pairs_analysis) * 100

print(f"\n[품질 분포 요약]")
print(f"  고품질 (>0.7): {high_quality_ratio:.1f}%")
print(f"  중품질 (0.5-0.7): {medium_quality_ratio:.1f}%")
print(f"  저품질 (<0.5): {low_quality_ratio:.1f}%")

print(f"\n[권장사항]")

if high_quality_ratio > 30:
    print("  ✅ 고품질 PAIR 비율이 높음 (30%+)")
    print("     → 0.7 이상만 선택하는 공격적 전략 가능")
elif high_quality_ratio > 15:
    print("  ✅ 고품질 PAIR 비율이 적절함 (15-30%)")
    print("     → 0.6-0.65 임계값 추천 (균형)")
else:
    print("  ⚠️  고품질 PAIR 비율이 낮음 (<15%)")
    print("     → 0.5 이상 포함하는 유연한 전략 필요")

if len(pairs_analysis) > 2700:
    top_2700_threshold = pairs_analysis['max_corr'].nlargest(2700).iloc[-1]
    top_2700_quality = pairs_analysis.nlargest(2700, 'max_corr')
    avg_corr_2700 = top_2700_quality['max_corr'].mean()
    avg_dtw_2700 = top_2700_quality['dtw_distance'].mean()

    print(f"\n[2,700개 선택 시]")
    print(f"  임계값: {top_2700_threshold:.4f}")
    print(f"  평균 상관계수: {avg_corr_2700:.4f}")
    print(f"  평균 DTW 거리: {avg_dtw_2700:.1f}")

    if avg_corr_2700 > 0.65:
        print(f"  ✅ 평균 품질 우수 (corr > 0.65)")
    elif avg_corr_2700 > 0.55:
        print(f"  ✅ 평균 품질 양호 (corr > 0.55)")
    else:
        print(f"  ⚠️  평균 품질 주의 필요 (corr < 0.55)")

print(f"\n[상관계수 vs DTW 관계]")
if corr_dtw_relationship < -0.5:
    print(f"  ✅ 강한 음의 상관 ({corr_dtw_relationship:.3f})")
    print(f"     → 상관계수만 사용해도 충분")
elif corr_dtw_relationship < -0.3:
    print(f"  ✅ 중간 음의 상관 ({corr_dtw_relationship:.3f})")
    print(f"     → 상관계수 우선, DTW 보조")
else:
    print(f"  ⚠️  약한 상관 ({corr_dtw_relationship:.3f})")
    print(f"     → DTW도 함께 고려 필요")

print(f"\n[사분면 분석 결과]")
if q1 > len(pairs_analysis) * 0.2:
    print(f"  ✅ Q1 (최고품질) 비율 높음 ({q1/len(pairs_analysis)*100:.1f}%)")
    print(f"     → 품질 좋은 데이터셋")
else:
    print(f"  ⚠️  Q1 (최고품질) 비율 낮음 ({q1/len(pairs_analysis)*100:.1f}%)")
    print(f"     → 필터링 전략 신중히 설정")

if q4 > len(pairs_analysis) * 0.2:
    print(f"  ⚠️  Q4 (저품질) 비율 높음 ({q4/len(pairs_analysis)*100:.1f}%)")
    print(f"     → 저품질 PAIR 제거 강화 필요")
else:
    print(f"  ✅ Q4 (저품질) 비율 낮음 ({q4/len(pairs_analysis)*100:.1f}%)")

print("\n" + "="*80)
print("✅ PAIR 품질 분포 분석 완료!")
print("="*80)
print("\n생성된 파일:")
print("  1. pair_quality_analysis.png - 9개 시각화")
print("  2. pair_quality_detailed.csv - 전체 PAIR 상세 정보")
print("  3. pair_quality_summary.csv - 요약 통계")
print("="*80)
```

```python
"""
무역 데이터 공행성 예측 모델 (17개 피처 최적화 버전)
🎯 피처 최적화: 33개 → 17개 (48% 감소)
📊 성능 유지: 91.8% 커버율
- 원본: 3,368개
- 목표: 2,700개 (약 80% 유지)
- 17개 피처 (핵심만 선택) + LGBM(0.7) + XGBoost(0.3) 앙상블
"""

import datetime
import pandas as pd
import numpy as np
from scipy import stats
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# 데이터 로드
# ============================================================================

print("="*80)
print("데이터 로드")
print("="*80)

monthly = (
    train
    .groupby(["item_id", "year", "month"], as_index=False)
    .agg({'value': 'sum', 'weight': 'sum'})
)

monthly["ym"] = pd.to_datetime(
    monthly["year"].astype(str) + "-" + monthly["month"].astype(str).str.zfill(2)
)

pivot = monthly.pivot(index="item_id", columns="ym", values="value").fillna(0.0)
pivot_weight = monthly.pivot(index="item_id", columns="ym", values="weight").fillna(0.0)

months_dt = sorted(pivot.columns.to_list())
pivot = pivot[months_dt]
pivot_weight = pivot_weight[months_dt]

print(f"✓ 데이터 로드 완료: {len(pivot)}개 품목, {len(months_dt)}개월")

# ============================================================================
# ⭐ HS2 매핑 생성
# ============================================================================

print("\n" + "="*80)
print("⭐ HS2 매핑 생성")
print("="*80)

item_to_hs4 = train.groupby('item_id')['hs4'].first().to_dict()
item_to_hs2 = {item: hs4 // 100 for item, hs4 in item_to_hs4.items()}

print(f"✓ {len(item_to_hs2)}개 품목의 HS2 매핑 생성")
print(f"✓ 총 {len(set(item_to_hs2.values()))}개 HS2 산업 분류")

# ============================================================================
# 품질 점수 계산
# ============================================================================

def calculate_quality_score_moderate(a_series, b_series, lag):
    """중간 품질 점수 계산"""

    scores = {}

    zero_ratio_a = (a_series == 0).sum() / len(a_series)
    zero_ratio_b = (b_series == 0).sum() / len(b_series)
    max_zero = max(zero_ratio_a, zero_ratio_b)

    if max_zero < 0.17:
        scores['zero'] = 10
    elif max_zero < 0.37:
        scores['zero'] = 7
    elif max_zero < 0.57:
        scores['zero'] = 4
    else:
        scores['zero'] = 0

    mean_a = np.mean(a_series)
    mean_b = np.mean(b_series)

    cv_a = np.std(a_series) / (mean_a + 1) if mean_a > 0 else 0
    cv_b = np.std(b_series) / (mean_b + 1) if mean_b > 0 else 0
    max_cv = max(cv_a, cv_b)

    if max_cv < 0.75:
        scores['cv'] = 10
    elif max_cv < 1.4:
        scores['cv'] = 7
    elif max_cv < 2.7:
        scores['cv'] = 4
    else:
        scores['cv'] = 0

    a_changes = np.abs(np.diff(a_series))
    b_changes = np.abs(np.diff(b_series))

    if len(a_changes) > 0 and len(b_changes) > 0:
        a_spike = a_changes.max() / (np.median(a_changes) + 1)
        b_spike = b_changes.max() / (np.median(b_changes) + 1)
        max_spike = max(a_spike, b_spike)

        if max_spike < 4.5:
            scores['spike'] = 10
        elif max_spike < 9.5:
            scores['spike'] = 7
        elif max_spike < 17:
            scores['spike'] = 4
        else:
            scores['spike'] = 0
    else:
        scores['spike'] = 5

    n = len(a_series)
    mid = n // 2

    try:
        if mid >= 12 and mid + lag < len(b_series):
            corr_first = np.corrcoef(
                a_series[:mid],
                b_series[lag:mid+lag]
            )[0, 1]

            corr_second = np.corrcoef(
                a_series[mid:],
                b_series[mid+lag:]
            )[0, 1]

            if not np.isnan(corr_first) and not np.isnan(corr_second):
                consistency = 1 - abs(corr_first - corr_second)

                if consistency > 0.52:
                    scores['stability'] = 15
                elif consistency > 0.35:
                    scores['stability'] = 10
                elif consistency > 0.22:
                    scores['stability'] = 6
                else:
                    scores['stability'] = 3
            else:
                scores['stability'] = 5
        else:
            scores['stability'] = 5
    except:
        scores['stability'] = 5

    if len(a_series) > 0 and len(b_series) > 0:
        x = np.arange(len(a_series))

        if np.std(a_series) > 0:
            slope_a, _ = np.polyfit(x, a_series, 1)
        else:
            slope_a = 0

        if np.std(b_series) > 0:
            slope_b, _ = np.polyfit(x, b_series, 1)
        else:
            slope_b = 0

        if (slope_a > 0 and slope_b > 0) or (slope_a < 0 and slope_b < 0):
            ratio = abs(slope_a) / (abs(slope_b) + 0.001)
            if 0.38 < ratio < 2.7:
                scores['trend'] = 10
            elif 0.25 < ratio < 4.2:
                scores['trend'] = 6
            else:
                scores['trend'] = 3
        elif slope_a == 0 or slope_b == 0:
            scores['trend'] = 5
        else:
            scores['trend'] = 2
    else:
        scores['trend'] = 4

    if len(a_changes) > lag and len(b_changes) > lag:
        direction_match = (np.sign(a_changes[:-lag]) == np.sign(b_changes[lag:])).mean()

        if direction_match > 0.64:
            scores['direction'] = 10
        elif direction_match > 0.54:
            scores['direction'] = 7
        elif direction_match > 0.44:
            scores['direction'] = 4
        else:
            scores['direction'] = 1
    else:
        scores['direction'] = 4

    try:
        if len(a_series) > lag:
            _, p_value = stats.pearsonr(
                a_series[:-lag] if lag > 0 else a_series,
                b_series[lag:] if lag > 0 else b_series
            )

            if p_value < 0.007:
                scores['significance'] = 10
            elif p_value < 0.025:
                scores['significance'] = 7
            elif p_value < 0.07:
                scores['significance'] = 4
            else:
                scores['significance'] = 1
        else:
            scores['significance'] = 4
    except:
        scores['significance'] = 4

    try:
        if len(b_series) > 1:
            autocorr = np.corrcoef(b_series[:-1], b_series[1:])[0, 1]

            if not np.isnan(autocorr):
                if 0.24 < autocorr < 0.86:
                    scores['autocorr'] = 10
                elif 0.09 < autocorr < 0.91:
                    scores['autocorr'] = 7
                else:
                    scores['autocorr'] = 3
            else:
                scores['autocorr'] = 4
        else:
            scores['autocorr'] = 4
    except:
        scores['autocorr'] = 4

    def count_outliers(series):
        q1 = np.percentile(series, 25)
        q3 = np.percentile(series, 75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outliers = ((series < lower) | (series > upper)).sum()
        return outliers / len(series)

    outlier_ratio_a = count_outliers(a_series)
    outlier_ratio_b = count_outliers(b_series)
    max_outlier = max(outlier_ratio_a, outlier_ratio_b)

    if max_outlier < 0.085:
        scores['outlier'] = 10
    elif max_outlier < 0.16:
        scores['outlier'] = 7
    elif max_outlier < 0.26:
        scores['outlier'] = 4
    else:
        scores['outlier'] = 1

    try:
        if len(b_series) >= 8:
            _, p_shapiro = stats.shapiro(b_series)

            if p_shapiro > 0.02:
                scores['normality'] = 5
            elif p_shapiro > 0.004:
                scores['normality'] = 3
            else:
                scores['normality'] = 1
        else:
            scores['normality'] = 2
    except:
        scores['normality'] = 2

    total_score = sum(scores.values())
    return total_score, scores


# ============================================================================
# 노이즈 제거
# ============================================================================

def remove_noise_pairs_flexible(pairs, pivot, ratio=0.80, min_score=0):
    """품질점수 기반 유연한 노이즈 제거"""

    print("\n" + "="*80)
    print(f"⭐ 유연한 노이즈 제거 시작 (상위 {ratio*100:.0f}% 유지)")
    print("="*80)
    print(f"원본: {len(pairs):,}개")

    quality_scores = []

    print("품질 평가 중...")
    for row in tqdm(pairs.itertuples(), total=len(pairs), desc="평가"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        if leader not in pivot.index or follower not in pivot.index:
            quality_scores.append(0)
            continue

        a_series = pivot.loc[leader].values
        b_series = pivot.loc[follower].values

        total_score, _ = calculate_quality_score_moderate(a_series, b_series, lag)
        quality_scores.append(total_score)

    pairs_with_score = pairs.copy()
    pairs_with_score['quality_score'] = quality_scores

    sorted_scores = sorted(quality_scores, reverse=True)

    cut_index = int(len(pairs) * ratio)
    threshold = max(min_score, sorted_scores[cut_index])

    pairs_clean = pairs_with_score[pairs_with_score['quality_score'] >= threshold].copy()

    removed = len(pairs) - len(pairs_clean)
    kept_ratio = len(pairs_clean) / len(pairs) * 100

    print(f"\n⭐ 노이즈 제거 완료!")
    print(f"유지된 pair: {len(pairs_clean):,}개 ({kept_ratio:.1f}%)")
    print(f"제거된 pair: {removed:,}개")
    print(f"임계값(점수): {threshold}")

    return pairs_clean.drop(columns=['quality_score'], errors='ignore')


# ============================================================================
# follower_hs2 추가
# ============================================================================

def add_follower_hs2(pairs, item_to_hs2):
    """follower_hs2 추가"""

    print("\n" + "="*80)
    print("⭐ follower_hs2 정보 추가")
    print("="*80)

    pairs_with_hs2 = pairs.copy()
    pairs_with_hs2['follower_hs2'] = pairs_with_hs2['following_item_id'].map(item_to_hs2)

    missing = pairs_with_hs2['follower_hs2'].isna().sum()
    if missing > 0:
        print(f"⚠️  HS2 매핑 실패: {missing}개 → 기본값 0 부여")
        pairs_with_hs2['follower_hs2'] = pairs_with_hs2['follower_hs2'].fillna(0)

    print(f"✓ follower_hs2 추가 완료: {len(pairs_with_hs2)}개 PAIR")

    print(f"\n[Follower HS2 분포 (상위 10개)]")
    hs2_counts = pairs_with_hs2['follower_hs2'].value_counts().head(10)
    for hs2, count in hs2_counts.items():
        print(f"  HS2 {int(hs2):02d}: {count}개 PAIR")

    return pairs_with_hs2


# ============================================================================
# ⭐ Weight 피처 (상위 2개만)
# ============================================================================

def create_weight_features_top2(value, weight, value_series):
    """
    Weight 피처 상위 2개만
    - b_weight_value_ratio (873.6)
    - b_weight_for_high_value (701.4)
    """
    positive_values = value_series[value_series > 0]

    if len(positive_values) > 0:
        value_median = np.median(positive_values)
    else:
        value_median = 0

    # 1. b_weight_for_high_value
    if value > value_median:
        weight_for_high_value = weight * 1.5
    else:
        weight_for_high_value = 0

    # 2. b_weight_value_ratio
    if value_median > 0:
        weight_value_ratio = weight * (value / value_median)
    else:
        weight_value_ratio = 0

    return {
        'weight_for_high_value': weight_for_high_value,
        'weight_value_ratio': weight_value_ratio
    }


# ============================================================================
# ⭐ 17개 피처 기반 학습 함수
# ============================================================================

def build_training_data_optimized(pivot, pivot_weight, pairs, months_dt):
    """
    17개 핵심 피처만 사용
    제거: Leader 전체, Lag 정보, Weight 하위 2개, 시간 하위 2개, 관계 피처 전체
    """
    months = months_dt
    n_months = len(months)
    rows = []

    print("\n학습 데이터 생성 중...")
    for row in tqdm(pairs.itertuples(index=False), desc="데이터 생성"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        follower_hs2 = float(row.follower_hs2)

        if leader not in pivot.index or follower not in pivot.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        for t in range(max(lag, 12), n_months - 1):

            b_t = b_series[t]
            b_t_1 = b_series[t - 1]
            b_t_2 = b_series[t - 2]
            b_t_12 = b_series[t - 12]
            b_t_plus_1 = b_series[t + 1]
            b_weight_t = b_weight_series[t]
            target_month = months[t + 1].month

            # Follower 변화
            b_diff = b_t - b_t_1
            b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
            b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
            b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)

            # Follower 통계
            b_roll_mean_3 = (b_t + b_series[t - 1] + b_series[t - 2]) / 3
            b_roll_mean_6 = b_series[t - 5:t + 1].mean()
            b_roll_std_3 = np.std([b_t, b_series[t - 1], b_series[t - 2]])
            b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
            b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)

            # Follower 추가
            recent_12 = b_series[max(0, t - 11):t + 1]
            b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

            # Weight 상위 2개
            wft = create_weight_features_top2(b_t, b_weight_t, b_series)

            rows.append({
                # Follower 기본 (3개)
                "b_t": b_t,
                "b_t_1": b_t_1,
                "b_t_12": b_t_12,

                # Follower 통계 (5개)
                "b_roll_mean_3": b_roll_mean_3,
                "b_roll_mean_6": b_roll_mean_6,
                "b_roll_std_3": b_roll_std_3,
                "b_cv": b_cv,
                "b_ma_ratio": b_ma_ratio,

                # Follower 변화 (4개)
                "b_diff": b_diff,
                "b_pct_change": b_pct_change,
                "b_yoy_growth": b_yoy_growth,
                "b_acceleration": b_acceleration,

                # Follower 추가 (1개)
                "b_percentile_rank": b_percentile_rank,

                # Weight 상위 2개
                "b_weight_value_ratio": wft['weight_value_ratio'],
                "b_weight_for_high_value": wft['weight_for_high_value'],

                # HS2 산업 (1개)
                "follower_hs2": follower_hs2,

                # 시간 (1개)
                "month": float(target_month),

                # 타겟
                "target": np.log1p(b_t_plus_1)
            })

    return pd.DataFrame(rows)


# ============================================================================
# ⭐ 17개 피처 기반 예측 함수
# ============================================================================

def predict_optimized(pivot, pivot_weight, pairs, model_lgb, model_xgb, months_dt):
    """17개 피처로 예측"""

    months = months_dt
    n_months = len(months)
    t_last = n_months - 1
    t_prev = n_months - 2
    preds = []
    target_month = months[-1].month + 1 if months[-1].month < 12 else 1

    print("\n예측 중...")
    for row in tqdm(pairs.itertuples(index=False), desc="예측"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        follower_hs2 = float(row.follower_hs2)

        if leader not in pivot.index or follower not in pivot.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        if t_last - lag < 0 or t_last < 12:
            continue

        b_t = b_series[t_last]
        b_t_1 = b_series[t_prev]
        b_t_2 = b_series[t_prev - 1]
        b_t_12 = b_series[t_last - 12]
        b_weight_t = b_weight_series[t_last]

        b_diff = b_t - b_t_1
        b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
        b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
        b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)

        b_roll_mean_3 = (b_t + b_series[t_last-1] + b_series[t_last-2]) / 3
        b_roll_mean_6 = b_series[t_last-5:t_last+1].mean()
        b_roll_std_3 = np.std([b_t, b_series[t_last-1], b_series[t_last-2]])
        b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
        b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)

        recent_12 = b_series[max(0, t_last - 11):t_last + 1]
        b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

        wft = create_weight_features_top2(b_t, b_weight_t, b_series)

        # ⭐ 17개 피처
        X_test = np.array([[
            # Follower 기본 (3개)
            b_t, b_t_1, b_t_12,
            # Follower 통계 (5개)
            b_roll_mean_3, b_roll_mean_6, b_roll_std_3, b_cv, b_ma_ratio,
            # Follower 변화 (4개)
            b_diff, b_pct_change, b_yoy_growth, b_acceleration,
            # Follower 추가 (1개)
            b_percentile_rank,
            # Weight 상위 2개
            wft['weight_value_ratio'], wft['weight_for_high_value'],
            # HS2 (1개)
            follower_hs2,
            # 시간 (1개)
            float(target_month)
        ]])

        pred_lgb = model_lgb.predict(X_test)[0]
        pred_xgb = model_xgb.predict(X_test)[0]

        pred_log = 0.7 * pred_lgb + 0.3 * pred_xgb

        y_pred = np.expm1(pred_log)
        y_pred = max(0.0, float(y_pred))
        y_pred = int(round(y_pred))

        preds.append({
            "leading_item_id": leader,
            "following_item_id": follower,
            "value": y_pred
        })

    return pd.DataFrame(preds)


# ============================================================================
# ⭐ MAIN 실행
# ============================================================================

if __name__ == "__main__":

    print("\n" + "="*80)
    print("⭐ 17개 피처 최적화 버전")
    print("="*80)
    print("📊 33개 → 17개 (48% 감소)")
    print("📈 성능 유지: 91.8%")
    print("🎯 LGBM(0.7) + XGBoost(0.3) 앙상블")
    print("="*80)

    # 1) 노이즈 제거
    pairs_for_model = remove_noise_pairs_flexible(
        pairs=pairs_with_dtw,
        pivot=pivot,
        ratio=0.80
    )

    print(f"\n최종 사용 쌍: {len(pairs_for_model)}개")

    # 2) follower_hs2 추가
    pairs_with_hs2 = add_follower_hs2(pairs_for_model, item_to_hs2)

    # 3) 학습 데이터 생성
    df_train_model = build_training_data_optimized(
        pivot, pivot_weight, pairs_with_hs2, months_dt
    )
    print(f"✓ 학습 데이터: {df_train_model.shape}")

    # ⭐ 17개 피처
    feature_cols = [
        # Follower 기본 (3개)
        'b_t', 'b_t_1', 'b_t_12',
        # Follower 통계 (5개)
        'b_roll_mean_3', 'b_roll_mean_6', 'b_roll_std_3', 'b_cv', 'b_ma_ratio',
        # Follower 변화 (4개)
        'b_diff', 'b_pct_change', 'b_yoy_growth', 'b_acceleration',
        # Follower 추가 (1개)
        'b_percentile_rank',
        # Weight 상위 2개
        'b_weight_value_ratio', 'b_weight_for_high_value',
        # HS2 (1개)
        'follower_hs2',
        # 시간 (1개)
        'month'
    ]

    print(f"\n⭐ 17개 핵심 피처:")
    print(f"  - Follower 기본: 3개")
    print(f"  - Follower 통계: 5개")
    print(f"  - Follower 변화: 4개")
    print(f"  - Follower 추가: 1개")
    print(f"  - Weight: 2개 (상위만)")
    print(f"  - HS2 산업: 1개")
    print(f"  - 시간: 1개")
    print(f"\n❌ 제거된 16개 피처:")
    print(f"  - Leader 전체: 4개")
    print(f"  - Leader-Follower 관계: 4개")
    print(f"  - Weight 하위: 2개")
    print(f"  - Follower 변화 하위: 1개")
    print(f"  - 시간 하위: 2개")
    print(f"  - Lag 정보: 2개")

    train_X = df_train_model[feature_cols].values
    train_y = df_train_model["target"].values

    print(f"\n학습 샘플: {len(train_X):,}, 피처: {len(feature_cols)}")

    # 4) LGBM 학습
    print("\n" + "="*80)
    print("⭐ LGBM 학습 시작")
    print("="*80)

    model_lgb = LGBMRegressor(
        n_estimators=700,
        learning_rate=0.01,
        max_depth=8,
        num_leaves=50,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=0.1,
        min_split_gain=0.01,
        random_state=42,
        verbose=-1,
        n_jobs=-1
    )

    model_lgb.fit(train_X, train_y)
    print("✓ LGBM 학습 완료")

    # 5) XGBoost 학습
    print("\n" + "="*80)
    print("⭐ XGBoost 학습 시작")
    print("="*80)

    model_xgb = XGBRegressor(
        n_estimators=900,
        learning_rate=0.02,
        max_depth=8,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=0.1,
        random_state=42,
        tree_method='hist',
        verbosity=0,
        n_jobs=-1
    )

    model_xgb.fit(train_X, train_y)
    print("✓ XGBoost 학습 완료")

    # 6) 피처 중요도 분석
    print("\n" + "="*80)
    print("⭐ 피처 중요도 분석 (17개)")
    print("="*80)

    importance_df = pd.DataFrame({
        'feature': feature_cols,
        'lgb_importance': model_lgb.feature_importances_,
        'xgb_importance': model_xgb.feature_importances_
    })

    importance_df['ensemble_importance'] = (
        0.7 * importance_df['lgb_importance'] +
        0.3 * importance_df['xgb_importance']
    )

    importance_df = importance_df.sort_values('ensemble_importance', ascending=False)

    print("\n[앙상블 평균 중요도 Top 17]")
    print(importance_df[['feature', 'ensemble_importance']].to_string(index=False))

    if 'follower_hs2' in importance_df['feature'].values:
        rank = importance_df[importance_df['feature'] == 'follower_hs2'].index[0] + 1
        imp = importance_df[importance_df['feature'] == 'follower_hs2']['ensemble_importance'].values[0]
        print(f"\n⭐ follower_hs2 순위: {rank}위 / 17개 (중요도: {imp:.2f})")

    date_str_feat = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    importance_df.to_csv(f'_feature_importance_17_{date_str_feat}.csv', index=False)
    print(f"\n✓ 피처 중요도 저장: _feature_importance_17_{date_str_feat}.csv")

    # 7) 예측
    print("\n" + "="*80)
    print("⭐ 예측 시작 (LGBM 0.7 + XGBoost 0.3)")
    print("="*80)

    submission = predict_optimized(
        pivot, pivot_weight, pairs_with_hs2, model_lgb, model_xgb, months_dt
    )

    print(f"\n✓ 예측 완료: {len(submission)}개")

    # 8) 저장
    date_str = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    submission.to_csv(f'_result_17features_{date_str}.csv', index=False)

    print("\n" + "="*80)
    print("🎉 17개 피처 최적화 버전 완료!")
    print("="*80)
    print(f"  📊 피처: 33개 → 17개 (48% 감소)")
    print(f"  📈 성능: 91.8% 유지")
    print(f"  🎯 PAIR: {len(pairs_with_hs2)}개")
    print(f"  💾 파일: _result_17features_{date_str}.csv")
    print("="*80)
```

```python
"""
무역 데이터 공행성 예측 모델 (2,700개 목표 최적화 + Ridge 메타 스태킹)
- 원본: 3,368개
- 목표: 2,700개 (약 80% 유지)
- 적절한 노이즈 제거 (중간 모드)
- 32개 피처 + LGBM/XGB + Ridge 메타 모델
"""

import datetime
import pandas as pd
import numpy as np
from scipy import stats
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from sklearn.linear_model import Ridge
from tqdm import tqdm

# ============================================================================
# 데이터 로드
# ============================================================================

print("="*80)
print("데이터 로드")
print("="*80)

monthly = (
    train
    .groupby(["item_id", "year", "month"], as_index=False)
    .agg({'value': 'sum', 'weight': 'sum'})
)

monthly["ym"] = pd.to_datetime(
    monthly["year"].astype(str) + "-" + monthly["month"].astype(str).str.zfill(2)
)

pivot = monthly.pivot(index="item_id", columns="ym", values="value").fillna(0.0)
pivot_weight = monthly.pivot(index="item_id", columns="ym", values="weight").fillna(0.0)
months_dt = pivot.columns.to_list()

print(f"✓ 데이터 로드 완료: {len(pivot)}개 품목, {len(months_dt)}개월")

# ============================================================================
# ⭐ 중간 품질 점수 계산 (2,700개 목표용)
# ============================================================================

def calculate_quality_score_moderate(a_series, b_series, lag):
    """
    중간 품질 점수 계산 (0-100점)
    - 원본 3,368개 → 목표 2,700개
    - 적절한 노이즈 제거 (80% 유지)
    """

    scores = {}

    # 1. 0 비율 (10점) - 중간
    zero_ratio_a = (a_series == 0).sum() / len(a_series)
    zero_ratio_b = (b_series == 0).sum() / len(b_series)
    max_zero = max(zero_ratio_a, zero_ratio_b)

    if max_zero < 0.17:
        scores['zero'] = 10
    elif max_zero < 0.37:
        scores['zero'] = 7
    elif max_zero < 0.57:
        scores['zero'] = 4
    else:
        scores['zero'] = 0

    # 2. 변동계수 (10점) - 중간
    cv_a = np.std(a_series) / (np.mean(a_series) + 1)
    cv_b = np.std(b_series) / (np.mean(b_series) + 1)
    max_cv = max(cv_a, cv_b)

    if max_cv < 0.75:
        scores['cv'] = 10
    elif max_cv < 1.4:
        scores['cv'] = 7
    elif max_cv < 2.7:
        scores['cv'] = 4
    else:
        scores['cv'] = 0

    # 3. 스파이크 (10점) - 중간
    a_changes = np.abs(np.diff(a_series))
    b_changes = np.abs(np.diff(b_series))

    if len(a_changes) > 0 and len(b_changes) > 0:
        a_spike = a_changes.max() / (np.median(a_changes) + 1)
        b_spike = b_changes.max() / (np.median(b_changes) + 1)
        max_spike = max(a_spike, b_spike)

        if max_spike < 4.5:
            scores['spike'] = 10
        elif max_spike < 9.5:
            scores['spike'] = 7
        elif max_spike < 17:
            scores['spike'] = 4
        else:
            scores['spike'] = 0
    else:
        scores['spike'] = 5

    # 4. 안정성 (15점) - 중간
    n = len(a_series)

    try:
        mid = n // 2

        if mid >= 12 and mid + lag < len(b_series):
            corr_first = np.corrcoef(
                a_series[:mid],
                b_series[lag:mid+lag]
            )[0, 1]

            corr_second = np.corrcoef(
                a_series[mid:],
                b_series[mid+lag:]
            )[0, 1]

            if not np.isnan(corr_first) and not np.isnan(corr_second):
                consistency = 1 - abs(corr_first - corr_second)

                if consistency > 0.52:
                    scores['stability'] = 15
                elif consistency > 0.35:
                    scores['stability'] = 10
                elif consistency > 0.22:
                    scores['stability'] = 6
                else:
                    scores['stability'] = 3
            else:
                scores['stability'] = 5
        else:
            scores['stability'] = 5
    except:
        scores['stability'] = 5

    # 5. 트렌드 일치 (10점) - 중간
    if len(a_series) > 0 and len(b_series) > 0:
        x = np.arange(len(a_series))

        if np.std(a_series) > 0:
            slope_a, _ = np.polyfit(x, a_series, 1)
        else:
            slope_a = 0

        if np.std(b_series) > 0:
            slope_b, _ = np.polyfit(x, b_series, 1)
        else:
            slope_b = 0

        if (slope_a > 0 and slope_b > 0) or (slope_a < 0 and slope_b < 0):
            ratio = abs(slope_a) / (abs(slope_b) + 1e-10)
            if 0.38 < ratio < 2.7:
                scores['trend'] = 10
            elif 0.25 < ratio < 4.2:
                scores['trend'] = 6
            else:
                scores['trend'] = 3
        elif slope_a == 0 or slope_b == 0:
            scores['trend'] = 5
        else:
            scores['trend'] = 2
    else:
        scores['trend'] = 4

    # 6. 방향 일치도 (10점) - 중간
    if len(a_changes) > lag and len(b_changes) > lag:
        direction_match = (np.sign(a_changes[:-lag]) == np.sign(b_changes[lag:])).mean()

        if direction_match > 0.64:
            scores['direction'] = 10
        elif direction_match > 0.54:
            scores['direction'] = 7
        elif direction_match > 0.44:
            scores['direction'] = 4
        else:
            scores['direction'] = 1
    else:
        scores['direction'] = 4

    # 7. 통계적 유의성 (10점) - 중간
    try:
        if len(a_series) > lag:
            _, p_value = stats.pearsonr(
                a_series[:-lag] if lag > 0 else a_series,
                b_series[lag:] if lag > 0 else b_series
            )

            if p_value < 0.007:
                scores['significance'] = 10
            elif p_value < 0.025:
                scores['significance'] = 7
            elif p_value < 0.07:
                scores['significance'] = 4
            else:
                scores['significance'] = 1
        else:
            scores['significance'] = 4
    except:
        scores['significance'] = 4

    # 8. 자기상관 (10점) - 중간
    try:
        if len(b_series) > 1:
            autocorr = np.corrcoef(b_series[:-1], b_series[1:])[0, 1]

            if 0.24 < autocorr < 0.86:
                scores['autocorr'] = 10
            elif 0.09 < autocorr < 0.91:
                scores['autocorr'] = 7
            else:
                scores['autocorr'] = 3
        else:
            scores['autocorr'] = 4
    except:
        scores['autocorr'] = 4

    # 9. 이상치 비율 (10점) - 중간
    def count_outliers(series):
        q1 = np.percentile(series, 25)
        q3 = np.percentile(series, 75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outliers = ((series < lower) | (series > upper)).sum()
        return outliers / len(series)

    outlier_ratio_a = count_outliers(a_series)
    outlier_ratio_b = count_outliers(b_series)
    max_outlier = max(outlier_ratio_a, outlier_ratio_b)

    if max_outlier < 0.085:
        scores['outlier'] = 10
    elif max_outlier < 0.16:
        scores['outlier'] = 7
    elif max_outlier < 0.26:
        scores['outlier'] = 4
    else:
        scores['outlier'] = 1

    # 10. 분포 정규성 (5점) - 중간
    try:
        if len(b_series) >= 8:
            _, p_shapiro = stats.shapiro(b_series)

            if p_shapiro > 0.02:
                scores['normality'] = 5
            elif p_shapiro > 0.004:
                scores['normality'] = 3
            else:
                scores['normality'] = 1
        else:
            scores['normality'] = 2
    except:
        scores['normality'] = 2

    total_score = sum(scores.values())

    return total_score, scores


# ============================================================================
# ⭐ 2,700개 목표 노이즈 제거
# ============================================================================

def remove_noise_pairs_for_2700(pairs, pivot, target_count=2700):
    """
    2,700개 목표 노이즈 제거

    원본: 3,368개
    목표: 2,700개 (약 80% 유지)
    전략: 적절한 노이즈 제거 (중간 모드)
    """

    print("\n" + "="*80)
    print("⭐ 2,700개 목표 노이즈 제거 (중간 모드)")
    print("="*80)
    print(f"원본: {len(pairs):,}개")
    print(f"목표: {target_count:,}개 (약 {target_count/len(pairs)*100:.0f}% 유지)")

    quality_scores = []
    detailed_scores = []

    print(f"\n품질 평가 중...")
    for row in tqdm(pairs.itertuples(), total=len(pairs), desc="평가"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        if leader not in pivot.index or follower not in pivot.index:
            quality_scores.append(0)
            detailed_scores.append({})
            continue

        a_series = pivot.loc[leader].values
        b_series = pivot.loc[follower].values

        total_score, scores = calculate_quality_score_moderate(a_series, b_series, lag)

        quality_scores.append(total_score)
        detailed_scores.append(scores)

    pairs_with_score = pairs.copy()
    pairs_with_score['quality_score'] = quality_scores

    sorted_scores = sorted(quality_scores, reverse=True)

    if len(sorted_scores) >= target_count:
        threshold = sorted_scores[target_count - 1]
        threshold = max(33, (threshold // 5) * 5)
    else:
        threshold = 33

    pairs_clean = pairs_with_score[
        pairs_with_score['quality_score'] >= threshold
    ].copy()

    pairs_for_model = pairs_clean.drop(columns=['quality_score'], errors='ignore')

    removed = len(pairs) - len(pairs_clean)
    removed_ratio = removed / len(pairs) * 100
    kept_ratio = len(pairs_clean) / len(pairs) * 100

    print(f"\n{'='*80}")
    print(f"✅ 노이즈 제거 완료!")
    print(f"{'='*80}")
    print(f"원본: {len(pairs):,}개")
    print(f"제거: {removed:,}개 ({removed_ratio:.1f}%)")
    print(f"유지: {len(pairs_clean):,}개 ({kept_ratio:.1f}%)")
    print(f"임계값: {threshold}점")

    return pairs_for_model


# ============================================================================
# 조건부 Weight 피처
# ============================================================================

def create_conditional_weight_features(value, weight, value_series):
    """조건부 Weight 피처 4개"""
    value_median = np.median(value_series[value_series > 0]) if (value_series > 0).any() else 0
    value_q75 = np.percentile(value_series[value_series > 0], 75) if (value_series > 0).any() else 0

    if value > value_median:
        weight_for_high_value = weight * 1.5
    else:
        weight_for_high_value = 0

    weight_value_ratio = weight * (value / (value_median + 1))
    exp_weighted = weight * (1 - np.exp(-value / (value_median + 1)))

    if value > value_q75:
        weight_importance = weight * 1.5
    elif value > value_median:
        weight_importance = weight * 1.0
    else:
        weight_importance = weight * 0.1

    return {
        'weight_for_high_value': weight_for_high_value,
        'weight_value_ratio': weight_value_ratio,
        'exp_weighted': exp_weighted,
        'weight_importance': weight_importance
    }


# ============================================================================
# 학습 데이터 구축 (32개 피처)
# ============================================================================

def build_training_data_optimized(pivot, pivot_weight, pairs, months_dt, exclude_recent_months=0):
    """
    32개 피처 학습 데이터 생성

    exclude_recent_months: 최근 N개월 제외 (메타 학습용)
    """
    months = months_dt
    n_months = len(months)

    # 메타 학습용 데이터 제외
    max_t = n_months - 1 - exclude_recent_months

    rows = []

    for row in pairs.itertuples(index=False):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        for t in range(max(lag, 12), max_t):
            b_t = b_series[t]
            b_t_1 = b_series[t - 1]
            b_t_2 = b_series[t - 2]
            b_t_12 = b_series[t - 12]
            a_t_lag = a_series[t - lag]
            b_t_plus_1 = b_series[t + 1]
            b_weight_t = b_weight_series[t]
            target_month = months[t + 1].month

            a_t_lag_diff = a_series[t - lag] - a_series[t - lag - 1]
            a_momentum = a_series[t - lag] / (a_series[t - lag - 3] + 1)
            a_roll_mean_3 = (a_series[t-lag] + a_series[t-lag-1] + a_series[t-lag-2]) / 3

            b_diff = b_t - b_t_1
            b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
            b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
            b_roll_mean_3 = (b_series[t] + b_series[t-1] + b_series[t-2]) / 3
            b_roll_mean_6 = (b_series[t-5:t+1]).mean()
            b_roll_std_3 = np.std([b_series[t], b_series[t-1], b_series[t-2]])

            ab_ratio = a_t_lag / (b_t + 1)
            ab_diff = a_t_lag - b_t
            corr_weighted_a = a_t_lag * corr

            b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
            b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
            b_yoy_ratio = b_t / (b_t_12 + 1)
            b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
            recent_12 = b_series[max(0, t-11):t+1]
            b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

            quarter = (target_month - 1) // 3 + 1
            is_year_end = 1 if target_month in [11, 12, 1] else 0
            lag_weight = 1 / (1 + lag)

            b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

            rows.append({
                "b_t": b_t, "b_t_1": b_t_1, "b_t_12": b_t_12, "a_t_lag": a_t_lag,
                "max_corr": corr, "best_lag": float(lag), "month": float(target_month),
                "a_t_lag_diff": a_t_lag_diff, "a_momentum": a_momentum, "a_roll_mean_3": a_roll_mean_3,
                "b_diff": b_diff, "b_pct_change": b_pct_change, "b_yoy_growth": b_yoy_growth,
                "b_roll_mean_3": b_roll_mean_3, "b_roll_mean_6": b_roll_mean_6, "b_roll_std_3": b_roll_std_3,
                "ab_ratio": ab_ratio, "ab_diff": ab_diff, "corr_weighted_a": corr_weighted_a,
                "b_cv": b_cv, "b_acceleration": b_acceleration, "b_yoy_ratio": b_yoy_ratio,
                "b_ma_ratio": b_ma_ratio, "b_percentile_rank": b_percentile_rank,
                "quarter": float(quarter), "is_year_end": is_year_end, "lag_weight": lag_weight,
                "b_weight_for_high_value": b_weight_features['weight_for_high_value'],
                "b_weight_value_ratio": b_weight_features['weight_value_ratio'],
                "b_exp_weighted": b_weight_features['exp_weighted'],
                "b_weight_importance": b_weight_features['weight_importance'],
                "target": b_t_plus_1,
            })

    return pd.DataFrame(rows)


# ============================================================================
# ⭐ 메타 학습 데이터 생성 (2025년 1-7월)
# ============================================================================

def build_meta_training_data(pivot, pivot_weight, pairs, months_dt, model_lgb, model_xgb,
                              feature_cols, meta_months=7):
    """
    2025년 최근 N개월 데이터로 메타 모델 학습용 데이터 생성

    Returns:
        meta_X: LGBM, XGB 예측값 (N, 2)
        meta_y: 실제 정답값 (N,)
    """
    months = months_dt
    n_months = len(months)

    # 메타 학습 시작 인덱스 (최근 meta_months+1개월)
    meta_start_idx = n_months - meta_months - 1

    meta_preds_lgb = []
    meta_preds_xgb = []
    meta_targets = []

    print(f"\n{'='*80}")
    print(f"⭐ 메타 학습 데이터 생성 (최근 {meta_months}개월)")
    print(f"{'='*80}")

    for row in tqdm(pairs.itertuples(index=False), desc="메타 데이터"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        # 최근 meta_months개월에 대해 예측
        for t in range(meta_start_idx, n_months - 1):
            if t - lag < 0 or t < 12:
                continue

            # 피처 생성 (기존과 동일)
            b_t = b_series[t]
            b_t_1 = b_series[t - 1]
            b_t_2 = b_series[t - 2]
            b_t_12 = b_series[t - 12]
            a_t_lag = a_series[t - lag]
            b_weight_t = b_weight_series[t]
            target_month = months[t + 1].month

            a_t_lag_diff = a_series[t - lag] - a_series[t - lag - 1]
            a_momentum = a_series[t - lag] / (a_series[t - lag - 3] + 1)
            a_roll_mean_3 = (a_series[t-lag] + a_series[t-lag-1] + a_series[t-lag-2]) / 3

            b_diff = b_t - b_t_1
            b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
            b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
            b_roll_mean_3 = (b_series[t] + b_series[t-1] + b_series[t-2]) / 3
            b_roll_mean_6 = b_series[t-5:t+1].mean()
            b_roll_std_3 = np.std([b_series[t], b_series[t-1], b_series[t-2]])

            ab_ratio = a_t_lag / (b_t + 1)
            ab_diff = a_t_lag - b_t
            corr_weighted_a = a_t_lag * corr

            b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
            b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
            b_yoy_ratio = b_t / (b_t_12 + 1)
            b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
            recent_12 = b_series[max(0, t-11):t+1]
            b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

            quarter = (target_month - 1) // 3 + 1
            is_year_end = 1 if target_month in [11, 12, 1] else 0
            lag_weight = 1 / (1 + lag)

            b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

            # 피처 배열 생성
            X_features = [
                b_t, b_t_1, b_t_12, a_t_lag, corr, float(lag), float(target_month),
                a_t_lag_diff, a_momentum, a_roll_mean_3,
                b_diff, b_pct_change, b_yoy_growth, b_roll_mean_3, b_roll_mean_6, b_roll_std_3,
                ab_ratio, ab_diff, corr_weighted_a,
                b_cv, b_acceleration, b_yoy_ratio, b_ma_ratio, b_percentile_rank,
                float(quarter), is_year_end, lag_weight,
                b_weight_features['weight_for_high_value'],
                b_weight_features['weight_value_ratio'],
                b_weight_features['exp_weighted'],
                b_weight_features['weight_importance']
            ]

            X_test = np.array([X_features])

            # 두 모델의 예측값
            pred_lgb = model_lgb.predict(X_test)[0]
            pred_xgb = model_xgb.predict(X_test)[0]

            # 실제 정답값
            actual = b_series[t + 1]

            meta_preds_lgb.append(pred_lgb)
            meta_preds_xgb.append(pred_xgb)
            meta_targets.append(actual)

    # (N, 2) 형태로 변환
    meta_X = np.column_stack([meta_preds_lgb, meta_preds_xgb])
    meta_y = np.array(meta_targets)

    print(f"✓ 메타 학습 샘플: {len(meta_y):,}개")

    return meta_X, meta_y


# ============================================================================
# ⭐ Ridge 메타 모델 학습
# ============================================================================

def train_meta_model(meta_X, meta_y):
    """
    Ridge 회귀로 메타 모델 학습

    alpha=1.0: L2 regularization 강도
    fit_intercept=False: 절편 없음 (순수 가중치만)
    positive=True: 음수 가중치 방지
    """
    print(f"\n{'='*80}")
    print(f"⭐ Ridge 메타 모델 학습")
    print(f"{'='*80}")

    meta_model = Ridge(alpha=1.0, fit_intercept=False, positive=True)
    meta_model.fit(meta_X, meta_y)

    # 학습된 가중치
    w_lgb, w_xgb = meta_model.coef_
    total_weight = w_lgb + w_xgb
    w_lgb_norm = w_lgb / total_weight
    w_xgb_norm = w_xgb / total_weight

    print(f"\n✓ Ridge 메타 모델 학습 완료!")
    print(f"{'='*80}")
    print(f"학습된 가중치:")
    print(f"  - LGBM: {w_lgb_norm:.4f} (기존: 0.7000)")
    print(f"  - XGB:  {w_xgb_norm:.4f} (기존: 0.3000)")
    print(f"{'='*80}")

    # 메타 학습 데이터에서의 성능 확인
    meta_pred = meta_model.predict(meta_X)
    mae = np.mean(np.abs(meta_y - meta_pred))
    rmse = np.sqrt(np.mean((meta_y - meta_pred) ** 2))

    print(f"\n메타 학습 데이터 성능:")
    print(f"  - MAE:  {mae:,.2f}")
    print(f"  - RMSE: {rmse:,.2f}")

    # 기존 0.7:0.3 가중치와 비교
    baseline_pred = 0.7 * meta_X[:, 0] + 0.3 * meta_X[:, 1]
    baseline_mae = np.mean(np.abs(meta_y - baseline_pred))
    baseline_rmse = np.sqrt(np.mean((meta_y - baseline_pred) ** 2))

    print(f"\n기존 0.7:0.3 가중치 성능:")
    print(f"  - MAE:  {baseline_mae:,.2f} (차이: {mae - baseline_mae:+.2f})")
    print(f"  - RMSE: {baseline_rmse:,.2f} (차이: {rmse - baseline_rmse:+.2f})")

    return meta_model, (w_lgb_norm, w_xgb_norm)


# ============================================================================
# 예측 함수 (Ridge 메타 모델 사용)
# ============================================================================

def predict_with_meta_model(pivot, pivot_weight, pairs, model_lgb, model_xgb,
                             meta_model, months_dt):
    """
    Ridge 메타 모델을 사용한 최종 예측
    """
    months = months_dt
    n_months = len(months)
    t_last = n_months - 1
    t_prev = n_months - 2
    preds = []
    target_month = months[-1].month + 1 if months[-1].month < 12 else 1

    # 디버깅 카운터
    total_pairs = 0
    skipped_not_in_pivot = 0
    skipped_condition = 0
    success_count = 0

    print(f"\n{'='*80}")
    print(f"⭐ Ridge 메타 모델 예측")
    print(f"{'='*80}")
    print(f"전체 pairs: {len(pairs)}개")
    print(f"예측 시점: t_last={t_last}, n_months={n_months}")

    for row in tqdm(pairs.itertuples(index=False), desc="예측"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        if t_last - lag < 0 or t_last < 12:
            continue

        b_t = b_series[t_last]
        b_t_1 = b_series[t_prev]
        b_t_2 = b_series[t_prev - 1]
        b_t_12 = b_series[t_last - 12]
        a_t_lag = a_series[t_last - lag]
        b_weight_t = b_weight_series[t_last]

        a_t_lag_diff = a_series[t_last - lag] - a_series[t_last - lag - 1]
        a_momentum = a_series[t_last - lag] / (a_series[t_last - lag - 3] + 1)
        a_roll_mean_3 = (a_series[t_last-lag] + a_series[t_last-lag-1] + a_series[t_last-lag-2]) / 3

        b_diff = b_t - b_t_1
        b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
        b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
        b_roll_mean_3 = (b_series[t_last] + b_series[t_last-1] + b_series[t_last-2]) / 3
        b_roll_mean_6 = b_series[t_last-5:t_last+1].mean()
        b_roll_std_3 = np.std([b_series[t_last], b_series[t_last-1], b_series[t_last-2]])

        ab_ratio = a_t_lag / (b_t + 1)
        ab_diff = a_t_lag - b_t
        corr_weighted_a = a_t_lag * corr

        b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
        b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
        b_yoy_ratio = b_t / (b_t_12 + 1)
        b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
        recent_12 = b_series[max(0, t_last-11):t_last+1]
        b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

        quarter = (target_month - 1) // 3 + 1
        is_year_end = 1 if target_month in [11, 12, 1] else 0
        lag_weight = 1 / (1 + lag)

        b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

        X_features = [
            b_t, b_t_1, b_t_12, a_t_lag, corr, float(lag), float(target_month),
            a_t_lag_diff, a_momentum, a_roll_mean_3,
            b_diff, b_pct_change, b_yoy_growth, b_roll_mean_3, b_roll_mean_6, b_roll_std_3,
            ab_ratio, ab_diff, corr_weighted_a,
            b_cv, b_acceleration, b_yoy_ratio, b_ma_ratio, b_percentile_rank,
            float(quarter), is_year_end, lag_weight,
            b_weight_features['weight_for_high_value'],
            b_weight_features['weight_value_ratio'],
            b_weight_features['exp_weighted'],
            b_weight_features['weight_importance']
        ]

        X_test = np.array([X_features])

        # LGBM, XGB 예측
        pred_lgb = model_lgb.predict(X_test)[0]
        pred_xgb = model_xgb.predict(X_test)[0]

        # Ridge 메타 모델로 최종 예측
        meta_input = np.array([[pred_lgb, pred_xgb]])
        y_pred = meta_model.predict(meta_input)[0]

        y_pred = max(0.0, float(y_pred))
        y_pred = int(round(y_pred))

        preds.append({
            "leading_item_id": leader,
            "following_item_id": follower,
            "value": y_pred,
        })

    return pd.DataFrame(preds)


# ============================================================================
# 메인 실행
# ============================================================================

if __name__ == "__main__":

    print("\n" + "="*80)
    print("⭐ Ridge 메타 스태킹 버전 (2,700개 목표)")
    print("="*80)

    # ⭐ 2,700개 목표 노이즈 제거
    pairs_for_model = remove_noise_pairs_for_2700(
        pairs=pairs_with_dtw,
        pivot=pivot,
        target_count=2700
    )

    print(f"\n최종 사용 쌍: {len(pairs_for_model)}개")

    feature_cols = [
        'b_t', 'b_t_1', 'b_t_12', 'a_t_lag', 'max_corr', 'best_lag', 'month',
        'a_t_lag_diff', 'a_momentum', 'a_roll_mean_3',
        'b_diff', 'b_pct_change', 'b_yoy_growth', 'b_roll_mean_3', 'b_roll_mean_6', 'b_roll_std_3',
        'ab_ratio', 'ab_diff', 'corr_weighted_a',
        'b_cv', 'b_acceleration', 'b_yoy_ratio', 'b_ma_ratio', 'b_percentile_rank',
        'quarter', 'is_year_end', 'lag_weight',
        'b_weight_for_high_value', 'b_weight_value_ratio', 'b_exp_weighted', 'b_weight_importance'
    ]

    # ============================================================================
    # 1단계: Base 모델 학습 (2022~2024 데이터, 최근 7개월 제외)
    # ============================================================================

    print("\n" + "="*80)
    print("1단계: Base 모델 학습 (LGBM, XGBoost)")
    print("="*80)
    print("학습 데이터: 2022~2024 + 2025년 초반 (최근 7개월 제외)")

    # 최근 7개월을 제외한 학습 데이터 생성
    df_train_base = build_training_data_optimized(
        pivot, pivot_weight, pairs_for_model, months_dt,
        exclude_recent_months=7  # 최근 7개월 제외
    )

    print(f"✓ Base 학습 데이터: {df_train_base.shape}")

    if not df_train_base.empty:
        train_X = df_train_base[feature_cols].values
        train_y = df_train_base["target"].values

        print(f"Base 학습 샘플: {len(train_X):,}, 피처: {len(feature_cols)}")

        # LGBM 학습
        model_lgb = LGBMRegressor(
            random_state=42, n_estimators=700, learning_rate=0.01, max_depth=8,
            num_leaves=50, min_child_samples=20, subsample=0.8, colsample_bytree=0.8,
            reg_alpha=0.1, reg_lambda=0.1, min_split_gain=0.01, verbose=-1
        )

        print("\nLGBM 학습 중...")
        model_lgb.fit(train_X, train_y)
        print("✓ LGBM 완료")

        # XGBoost 학습
        model_xgb = XGBRegressor(
            random_state=42, n_estimators=900, learning_rate=0.02, max_depth=8,
            subsample=0.8, colsample_bytree=0.8, reg_alpha=0.1, reg_lambda=0.1,
            tree_method='hist', verbosity=0
        )

        print("XGBoost 학습 중...")
        model_xgb.fit(train_X, train_y)
        print("✓ XGBoost 완료")

        # ============================================================================
        # 2단계: 메타 모델 학습 (2025년 최근 7개월)
        # ============================================================================

        # 메타 학습 데이터 생성
        meta_X, meta_y = build_meta_training_data(
            pivot, pivot_weight, pairs_for_model, months_dt,
            model_lgb, model_xgb, feature_cols, meta_months=7
        )

        # Ridge 메타 모델 학습
        meta_model, weights = train_meta_model(meta_X, meta_y)

        # ============================================================================
        # 3단계: 최종 예측 (Ridge 메타 모델 사용)
        # ============================================================================

        submission = predict_with_meta_model(
            pivot, pivot_weight, pairs_for_model,
            model_lgb, model_xgb, meta_model, months_dt
        )

        print(f"\n✓ 예측 완료: {len(submission)}개")

        # 저장
        date_str = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

        submission.to_csv('sy_2700_ridge_meta_{date_str}.csv', index=False)

        print(f"\n{'='*80}")

        print(f"{'='*80}")
        print("\n✅ Ridge 메타 스태킹 버전 완료!")
        print("="*80)
        print("주요 개선사항:")
        print("  ✓ 원본: 3,368개 → 목표: 2,700개")
        print("  ✓ 중간 기준 (적절한 노이즈 제거)")
        print("  ✓ 32개 피처")
        print(f"  ✓ LGBM + XGB + Ridge 메타 스태킹")
        print(f"  ✓ 학습된 가중치: LGBM {weights[0]:.4f}, XGB {weights[1]:.4f}")
        print("="*80)
    else:
        print("\n❌ 학습 데이터가 비어있습니다!")
```

#왜 Ridge 메타 스태킹이 성능이 떨어졌을까?
베이스 모델이 보는 양 차이
7개월간 보고 있기 때문에 만약 2025-01~07 구간에서 우연히 XGB가 좀 더 잘 맞았고 Ridge가 XGB 비중을 0.62로 높게줬는데 실제 08월에는 LBGM이 더 잘 맞는 구간이 많다면 0.7/0.3 고정이 오히려 더 보수적인 평균 역할을 하면서 잘 버티는 것



```python
"""
무역 데이터 공행성 예측 모델 (2,700개 목표 최적화)
- 원본: 3,368개
- 목표: 2,700개 (약 80% 유지)
- 적절한 노이즈 제거 (중간 모드)
- 32개 피처 + LGBM/XGB 앙상블
"""

import datetime
import pandas as pd
import numpy as np
from scipy import stats
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from tqdm import tqdm

# ============================================================================
# 데이터 로드
# ============================================================================

print("="*80)
print("데이터 로드")
print("="*80)

monthly = (
    train
    .groupby(["item_id", "year", "month"], as_index=False)
    .agg({'value': 'sum', 'weight': 'sum'})
)

monthly["ym"] = pd.to_datetime(
    monthly["year"].astype(str) + "-" + monthly["month"].astype(str).str.zfill(2)
)

pivot = monthly.pivot(index="item_id", columns="ym", values="value").fillna(0.0)
pivot_weight = monthly.pivot(index="item_id", columns="ym", values="weight").fillna(0.0)
months_dt = pivot.columns.to_list()

print(f"✓ 데이터 로드 완료: {len(pivot)}개 품목, {len(months_dt)}개월")

# ============================================================================
# ⭐ 중간 품질 점수 계산 (2,700개 목표용)
# ============================================================================

def calculate_quality_score_moderate(a_series, b_series, lag):
    """
    중간 품질 점수 계산 (0-100점)
    - 원본 3,368개 → 목표 2,700개
    - 적절한 노이즈 제거 (80% 유지)
    """

    scores = {}

    # 1. 0 비율 (10점) - 중간
    zero_ratio_a = (a_series == 0).sum() / len(a_series)
    zero_ratio_b = (b_series == 0).sum() / len(b_series)
    max_zero = max(zero_ratio_a, zero_ratio_b)

    if max_zero < 0.17:
        scores['zero'] = 10
    elif max_zero < 0.37:
        scores['zero'] = 7
    elif max_zero < 0.57:  # 57%까지 허용
        scores['zero'] = 4
    else:
        scores['zero'] = 0

    # 2. 변동계수 (10점) - 중간
    cv_a = np.std(a_series) / (np.mean(a_series) + 1)
    cv_b = np.std(b_series) / (np.mean(b_series) + 1)
    max_cv = max(cv_a, cv_b)

    if max_cv < 0.75:
        scores['cv'] = 10
    elif max_cv < 1.4:
        scores['cv'] = 7
    elif max_cv < 2.7:  # 2.7까지 허용
        scores['cv'] = 4
    else:
        scores['cv'] = 0

    # 3. 스파이크 (10점) - 중간
    a_changes = np.abs(np.diff(a_series))
    b_changes = np.abs(np.diff(b_series))

    if len(a_changes) > 0 and len(b_changes) > 0:
        a_spike = a_changes.max() / (np.median(a_changes) + 1)
        b_spike = b_changes.max() / (np.median(b_changes) + 1)
        max_spike = max(a_spike, b_spike)

        if max_spike < 4.5:
            scores['spike'] = 10
        elif max_spike < 9.5:
            scores['spike'] = 7
        elif max_spike < 17:  # 17까지 허용
            scores['spike'] = 4
        else:
            scores['spike'] = 0
    else:
        scores['spike'] = 5

    # 4. 안정성 (15점) - 중간
    n = len(a_series)

    try:
        mid = n // 2

        if mid >= 12 and mid + lag < len(b_series):
            corr_first = np.corrcoef(
                a_series[:mid],
                b_series[lag:mid+lag]
            )[0, 1]

            corr_second = np.corrcoef(
                a_series[mid:],
                b_series[mid+lag:]
            )[0, 1]

            if not np.isnan(corr_first) and not np.isnan(corr_second):
                consistency = 1 - abs(corr_first - corr_second)

                if consistency > 0.52:  # 중간
                    scores['stability'] = 15
                elif consistency > 0.35:
                    scores['stability'] = 10
                elif consistency > 0.22:
                    scores['stability'] = 6
                else:
                    scores['stability'] = 3
            else:
                scores['stability'] = 5
        else:
            scores['stability'] = 5
    except:
        scores['stability'] = 5

    # 5. 트렌드 일치 (10점) - 중간
    if len(a_series) > 0 and len(b_series) > 0:
        x = np.arange(len(a_series))

        if np.std(a_series) > 0:
            slope_a, _ = np.polyfit(x, a_series, 1)
        else:
            slope_a = 0

        if np.std(b_series) > 0:
            slope_b, _ = np.polyfit(x, b_series, 1)
        else:
            slope_b = 0

        if (slope_a > 0 and slope_b > 0) or (slope_a < 0 and slope_b < 0):
            ratio = abs(slope_a) / (abs(slope_b) + 1e-10)
            if 0.38 < ratio < 2.7:  # 중간
                scores['trend'] = 10
            elif 0.25 < ratio < 4.2:
                scores['trend'] = 6
            else:
                scores['trend'] = 3
        elif slope_a == 0 or slope_b == 0:
            scores['trend'] = 5
        else:
            scores['trend'] = 2
    else:
        scores['trend'] = 4

    # 6. 방향 일치도 (10점) - 중간
    if len(a_changes) > lag and len(b_changes) > lag:
        direction_match = (np.sign(a_changes[:-lag]) == np.sign(b_changes[lag:])).mean()

        if direction_match > 0.64:  # 중간
            scores['direction'] = 10
        elif direction_match > 0.54:
            scores['direction'] = 7
        elif direction_match > 0.44:
            scores['direction'] = 4
        else:
            scores['direction'] = 1
    else:
        scores['direction'] = 4

    # 7. 통계적 유의성 (10점) - 중간
    try:
        if len(a_series) > lag:
            _, p_value = stats.pearsonr(
                a_series[:-lag] if lag > 0 else a_series,
                b_series[lag:] if lag > 0 else b_series
            )

            if p_value < 0.007:  # 중간
                scores['significance'] = 10
            elif p_value < 0.025:
                scores['significance'] = 7
            elif p_value < 0.07:
                scores['significance'] = 4
            else:
                scores['significance'] = 1
        else:
            scores['significance'] = 4
    except:
        scores['significance'] = 4

    # 8. 자기상관 (10점) - 중간
    try:
        if len(b_series) > 1:
            autocorr = np.corrcoef(b_series[:-1], b_series[1:])[0, 1]

            if 0.24 < autocorr < 0.86:  # 중간
                scores['autocorr'] = 10
            elif 0.09 < autocorr < 0.91:
                scores['autocorr'] = 7
            else:
                scores['autocorr'] = 3
        else:
            scores['autocorr'] = 4
    except:
        scores['autocorr'] = 4

    # 9. 이상치 비율 (10점) - 중간
    def count_outliers(series):
        q1 = np.percentile(series, 25)
        q3 = np.percentile(series, 75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outliers = ((series < lower) | (series > upper)).sum()
        return outliers / len(series)

    outlier_ratio_a = count_outliers(a_series)
    outlier_ratio_b = count_outliers(b_series)
    max_outlier = max(outlier_ratio_a, outlier_ratio_b)

    if max_outlier < 0.085:
        scores['outlier'] = 10
    elif max_outlier < 0.16:
        scores['outlier'] = 7
    elif max_outlier < 0.26:  # 26%까지 허용
        scores['outlier'] = 4
    else:
        scores['outlier'] = 1

    # 10. 분포 정규성 (5점) - 중간
    try:
        if len(b_series) >= 8:
            _, p_shapiro = stats.shapiro(b_series)

            if p_shapiro > 0.02:  # 중간
                scores['normality'] = 5
            elif p_shapiro > 0.004:
                scores['normality'] = 3
            else:
                scores['normality'] = 1
        else:
            scores['normality'] = 2
    except:
        scores['normality'] = 2

    total_score = sum(scores.values())

    return total_score, scores


# ============================================================================
# ⭐ 2,700개 목표 노이즈 제거
# ============================================================================

def remove_noise_pairs_for_2700(pairs, pivot, target_count=2700):
    """
    2,700개 목표 노이즈 제거

    원본: 3,368개
    목표: 2,700개 (약 80% 유지)
    전략: 적절한 노이즈 제거 (중간 모드)
    """

    print("\n" + "="*80)
    print("⭐ 2,700개 목표 노이즈 제거 (중간 모드)")
    print("="*80)
    print(f"원본: {len(pairs):,}개")
    print(f"목표: {target_count:,}개 (약 {target_count/len(pairs)*100:.0f}% 유지)")

    quality_scores = []
    detailed_scores = []

    # 품질 점수 계산 (중간 기준)
    print(f"\n품질 평가 중...")
    for row in tqdm(pairs.itertuples(), total=len(pairs), desc="평가"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        if leader not in pivot.index or follower not in pivot.index:
            quality_scores.append(0)
            detailed_scores.append({})
            continue

        a_series = pivot.loc[leader].values
        b_series = pivot.loc[follower].values

        total_score, scores = calculate_quality_score_moderate(a_series, b_series, lag)

        quality_scores.append(total_score)
        detailed_scores.append(scores)

    # 품질 점수 추가
    pairs_with_score = pairs.copy()
    pairs_with_score['quality_score'] = quality_scores

    # 목표 개수 기반 임계값 계산
    sorted_scores = sorted(quality_scores, reverse=True)

    if len(sorted_scores) >= target_count:
        # 목표 개수번째 점수를 임계값으로
        threshold = sorted_scores[target_count - 1]
        # 5점 단위로 내림
        threshold = max(33, (threshold // 5) * 5)  # 최소 33점
    else:
        threshold = 33  # 최소 임계값

    # 필터링
    pairs_clean = pairs_with_score[
        pairs_with_score['quality_score'] >= threshold
    ].copy()

    # 품질 점수 컬럼 제거
    pairs_for_model = pairs_clean.drop(columns=['quality_score'], errors='ignore')

    # 통계
    removed = len(pairs) - len(pairs_clean)
    removed_ratio = removed / len(pairs) * 100
    kept_ratio = len(pairs_clean) / len(pairs) * 100

    print(f"\n{'='*80}")
    print(f"✅ 노이즈 제거 완료!")
    print(f"{'='*80}")
    print(f"원본: {len(pairs):,}개")
    print(f"제거: {removed:,}개 ({removed_ratio:.1f}%)")
    print(f"유지: {len(pairs_clean):,}개 ({kept_ratio:.1f}%)")
    print(f"임계값: {threshold}점")

    # 점수 분포
    print(f"\n품질 점수 분포:")
    print(f"  - 평균: {np.mean(quality_scores):.1f}점")
    print(f"  - 중앙값: {np.median(quality_scores):.1f}점")
    print(f"  - 최소: {np.min(quality_scores):.1f}점")
    print(f"  - 최대: {np.max(quality_scores):.1f}점")

    # 선택된 쌍의 점수
    selected_scores = pairs_clean['quality_score'].values
    print(f"\n선택된 쌍의 점수:")
    print(f"  - 평균: {np.mean(selected_scores):.1f}점")
    print(f"  - 최소: {np.min(selected_scores):.1f}점")

    # 구간별 분포
    print(f"\n점수 구간별 분포:")
    bins = [0, 30, 40, 50, 60, 70, 80, 90, 100]
    for i in range(len(bins)-1):
        total_count = sum((np.array(quality_scores) >= bins[i]) & (np.array(quality_scores) < bins[i+1]))
        selected_count = sum((selected_scores >= bins[i]) & (selected_scores < bins[i+1]))
        print(f"  {bins[i]:3d}-{bins[i+1]:3d}점: 전체 {total_count:4,}개, 선택 {selected_count:4,}개")

    # 기준별 평균 점수
    print(f"\n기준별 평균 점수 (선택된 쌍):")
    criteria_names = {
        'zero': '0 비율',
        'cv': '변동계수',
        'spike': '스파이크',
        'stability': '안정성',
        'trend': '트렌드 일치',
        'direction': '방향 일치도',
        'significance': '통계적 유의성',
        'autocorr': '자기상관',
        'outlier': '이상치 비율',
        'normality': '정규성'
    }

    selected_indices = pairs_with_score[pairs_with_score['quality_score'] >= threshold].index
    selected_detailed = [detailed_scores[i] for i in selected_indices if i < len(detailed_scores)]

    for key, name in criteria_names.items():
        scores = [s.get(key, 0) for s in selected_detailed if s]
        if scores:
            if key == 'stability':
                max_score = 15
            elif key == 'normality':
                max_score = 5
            else:
                max_score = 10

            print(f"  {name:15s}: 평균 {np.mean(scores):5.1f}점 (최대 {max_score}점)")

    return pairs_for_model


# ============================================================================
# 조건부 Weight 피처
# ============================================================================

def create_conditional_weight_features(value, weight, value_series):
    """조건부 Weight 피처 4개"""
    value_median = np.median(value_series[value_series > 0]) if (value_series > 0).any() else 0
    value_q75 = np.percentile(value_series[value_series > 0], 75) if (value_series > 0).any() else 0

    if value > value_median:
        weight_for_high_value = weight * 1.5
    else:
        weight_for_high_value = 0

    weight_value_ratio = weight * (value / (value_median + 1))
    exp_weighted = weight * (1 - np.exp(-value / (value_median + 1)))

    if value > value_q75:
        weight_importance = weight * 1.5
    elif value > value_median:
        weight_importance = weight * 1.0
    else:
        weight_importance = weight * 0.1

    return {
        'weight_for_high_value': weight_for_high_value,
        'weight_value_ratio': weight_value_ratio,
        'exp_weighted': exp_weighted,
        'weight_importance': weight_importance
    }


# ============================================================================
# 학습 데이터 구축 (32개 피처)
# ============================================================================

def build_training_data_optimized(pivot, pivot_weight, pairs, months_dt):
    """32개 피처 학습 데이터 생성"""
    months = months_dt
    n_months = len(months)
    rows = []

    for row in pairs.itertuples(index=False):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        for t in range(max(lag, 12), n_months - 1):
            b_t = b_series[t]
            b_t_1 = b_series[t - 1]
            b_t_2 = b_series[t - 2]
            b_t_12 = b_series[t - 12]
            a_t_lag = a_series[t - lag]
            b_t_plus_1 = b_series[t + 1]
            b_weight_t = b_weight_series[t]
            target_month = months[t + 1].month

            a_t_lag_diff = a_series[t - lag] - a_series[t - lag - 1]
            a_momentum = a_series[t - lag] / (a_series[t - lag - 3] + 1)
            a_roll_mean_3 = (a_series[t-lag] + a_series[t-lag-1] + a_series[t-lag-2]) / 3

            b_diff = b_t - b_t_1
            b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
            b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
            b_roll_mean_3 = (b_series[t] + b_series[t-1] + b_series[t-2]) / 3
            b_roll_mean_6 = (b_series[t-5:t+1]).mean()
            b_roll_std_3 = np.std([b_series[t], b_series[t-1], b_series[t-2]])

            ab_ratio = a_t_lag / (b_t + 1)
            ab_diff = a_t_lag - b_t
            corr_weighted_a = a_t_lag * corr

            b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
            b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
            b_yoy_ratio = b_t / (b_t_12 + 1)
            b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
            recent_12 = b_series[max(0, t-11):t+1]
            b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

            quarter = (target_month - 1) // 3 + 1
            is_year_end = 1 if target_month in [11, 12, 1] else 0
            lag_weight = 1 / (1 + lag)

            b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

            rows.append({
                "b_t": b_t, "b_t_1": b_t_1, "b_t_12": b_t_12, "a_t_lag": a_t_lag,
                "max_corr": corr, "best_lag": float(lag), "month": float(target_month),
                "a_t_lag_diff": a_t_lag_diff, "a_momentum": a_momentum, "a_roll_mean_3": a_roll_mean_3,
                "b_diff": b_diff, "b_pct_change": b_pct_change, "b_yoy_growth": b_yoy_growth,
                "b_roll_mean_3": b_roll_mean_3, "b_roll_mean_6": b_roll_mean_6, "b_roll_std_3": b_roll_std_3,
                "ab_ratio": ab_ratio, "ab_diff": ab_diff, "corr_weighted_a": corr_weighted_a,
                "b_cv": b_cv, "b_acceleration": b_acceleration, "b_yoy_ratio": b_yoy_ratio,
                "b_ma_ratio": b_ma_ratio, "b_percentile_rank": b_percentile_rank,
                "quarter": float(quarter), "is_year_end": is_year_end, "lag_weight": lag_weight,
                "b_weight_for_high_value": b_weight_features['weight_for_high_value'],
                "b_weight_value_ratio": b_weight_features['weight_value_ratio'],
                "b_exp_weighted": b_weight_features['exp_weighted'],
                "b_weight_importance": b_weight_features['weight_importance'],
                "target": np.log1p(b_t_plus_1),
            })

    return pd.DataFrame(rows)


# ============================================================================
# 예측 함수
# ============================================================================

def predict_optimized(pivot, pivot_weight, pairs, model_lgb, model_xgb, months_dt):
    """32개 피처 예측 + 앙상블"""
    months = months_dt
    n_months = len(months)
    t_last = n_months - 1
    t_prev = n_months - 2
    preds = []
    target_month = months[-1].month + 1 if months[-1].month < 12 else 1

    for row in tqdm(pairs.itertuples(index=False), desc="예측"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        if t_last - lag < 0 or t_last < 12:
            continue

        b_t = b_series[t_last]
        b_t_1 = b_series[t_prev]
        b_t_2 = b_series[t_prev - 1]
        b_t_12 = b_series[t_last - 12]
        a_t_lag = a_series[t_last - lag]
        b_weight_t = b_weight_series[t_last]

        a_t_lag_diff = a_series[t_last - lag] - a_series[t_last - lag - 1]
        a_momentum = a_series[t_last - lag] / (a_series[t_last - lag - 3] + 1)
        a_roll_mean_3 = (a_series[t_last-lag] + a_series[t_last-lag-1] + a_series[t_last-lag-2]) / 3

        b_diff = b_t - b_t_1
        b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
        b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
        b_roll_mean_3 = (b_series[t_last] + b_series[t_last-1] + b_series[t_last-2]) / 3
        b_roll_mean_6 = b_series[t_last-5:t_last+1].mean()
        b_roll_std_3 = np.std([b_series[t_last], b_series[t_last-1], b_series[t_last-2]])

        ab_ratio = a_t_lag / (b_t + 1)
        ab_diff = a_t_lag - b_t
        corr_weighted_a = a_t_lag * corr

        b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
        b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
        b_yoy_ratio = b_t / (b_t_12 + 1)
        b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
        recent_12 = b_series[max(0, t_last-11):t_last+1]
        b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

        quarter = (target_month - 1) // 3 + 1
        is_year_end = 1 if target_month in [11, 12, 1] else 0
        lag_weight = 1 / (1 + lag)

        b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

        X_test = np.array([[
            b_t, b_t_1, b_t_12, a_t_lag, corr, float(lag), float(target_month),
            a_t_lag_diff, a_momentum, a_roll_mean_3,
            b_diff, b_pct_change, b_yoy_growth, b_roll_mean_3, b_roll_mean_6, b_roll_std_3,
            ab_ratio, ab_diff, corr_weighted_a,
            b_cv, b_acceleration, b_yoy_ratio, b_ma_ratio, b_percentile_rank,
            float(quarter), is_year_end, lag_weight,
            b_weight_features['weight_for_high_value'],
            b_weight_features['weight_value_ratio'],
            b_weight_features['exp_weighted'],
            b_weight_features['weight_importance']
        ]])

        pred_lgb = model_lgb.predict(X_test)[0]
        pred_xgb = model_xgb.predict(X_test)[0]
        y_pred_log = 0.7 * pred_lgb + 0.3 * pred_xgb
        y_pred = np.expm1(y_pred_log)
        y_pred = max(0.0, float(y_pred))
        y_pred = int(round(y_pred))

        preds.append({
            "leading_item_id": leader,
            "following_item_id": follower,
            "value": y_pred,
        })

    return pd.DataFrame(preds)


# ============================================================================
# 메인 실행
# ============================================================================

if __name__ == "__main__":

    print("\n" + "="*80)
    print("⭐ 2,700개 목표 최적화 버전")
    print("="*80)

    # ⭐ 2,700개 목표 노이즈 제거
    pairs_for_model = remove_noise_pairs_for_2700(
        pairs=pairs_with_dtw,
        pivot=pivot,
        target_count=2700      # ⭐ 목표: 2,700개
    )

    print(f"\n최종 사용 쌍: {len(pairs_for_model)}개")

    # 학습
    print("\n" + "="*80)
    print("학습 데이터 생성")
    print("="*80)

    df_train_model = build_training_data_optimized(pivot, pivot_weight, pairs_for_model, months_dt)
    print(f"✓ 학습 데이터: {df_train_model.shape}")

    feature_cols = [
        'b_t', 'b_t_1', 'b_t_12', 'a_t_lag', 'max_corr', 'best_lag', 'month',
        'a_t_lag_diff', 'a_momentum', 'a_roll_mean_3',
        'b_diff', 'b_pct_change', 'b_yoy_growth', 'b_roll_mean_3', 'b_roll_mean_6', 'b_roll_std_3',
        'ab_ratio', 'ab_diff', 'corr_weighted_a',
        'b_cv', 'b_acceleration', 'b_yoy_ratio', 'b_ma_ratio', 'b_percentile_rank',
        'quarter', 'is_year_end', 'lag_weight',
        'b_weight_for_high_value', 'b_weight_value_ratio', 'b_exp_weighted', 'b_weight_importance'
    ]

    if not df_train_model.empty:
        print("\n" + "="*80)
        print("모델 학습")
        print("="*80)

        train_X = df_train_model[feature_cols].values
        train_y = df_train_model["target"].values

        print(f"학습 샘플: {len(train_X):,}, 피처: {len(feature_cols)}")

        model_lgb = LGBMRegressor(
            random_state=42, n_estimators=700, learning_rate=0.01, max_depth=8,
            num_leaves=50, min_child_samples=20, subsample=0.8, colsample_bytree=0.8,
            reg_alpha=0.1, reg_lambda=0.1, min_split_gain=0.01
        )

        print("\nLGBM 학습 중...")
        model_lgb.fit(train_X, train_y)
        print("✓ LGBM 완료")

        model_xgb = XGBRegressor(
            random_state=42, n_estimators=900, learning_rate=0.02, max_depth=8,
            subsample=0.8, colsample_bytree=0.8, reg_alpha=0.1, reg_lambda=0.1,
            tree_method='hist'
        )

        print("XGBoost 학습 중...")
        model_xgb.fit(train_X, train_y)
        print("✓ XGBoost 완료")

        print("\n" + "="*80)
        print("예측")
        print("="*80)

        submission = predict_optimized(pivot, pivot_weight, pairs_for_model, model_lgb, model_xgb, months_dt)
        print(f"\n✓ 예측 완료: {len(submission)}개")

        date_str = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

        submission.to_csv(f'_result_sy_2700_moderate_{date_str}.csv', index=False)


        print("\n✅ 2,700개 목표 버전 완료!")
        print("  - 원본: 3,368개")
        print("  - 목표: 2,700개")
        print("  - 중간 기준 (적절한 노이즈 제거)")
        print("  - 32개 피처")
        print("  - LGBM + XGB 앙상블")
        print("="*80)
```

```python
"""
무역 데이터 공행성 예측 모델 (2,700개 목표 최적화)
- 원본: 3,368개
- 목표: 2,700개 (약 80% 유지)
- 적절한 노이즈 제거 (중간 모드)
- 32개 피처 + LGBM/XGB 앙상블 (+ seed 앙상블)
"""

import datetime
import pandas as pd
import numpy as np
from scipy import stats
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from tqdm import tqdm

# ============================================================================
# 데이터 로드
# ============================================================================

print("="*80)
print("데이터 로드")
print("="*80)

monthly = (
    train
    .groupby(["item_id", "year", "month"], as_index=False)
    .agg({'value': 'sum', 'weight': 'sum'})
)

monthly["ym"] = pd.to_datetime(
    monthly["year"].astype(str) + "-" + monthly["month"].astype(str).str.zfill(2)
)

pivot = monthly.pivot(index="item_id", columns="ym", values="value").fillna(0.0)
pivot_weight = monthly.pivot(index="item_id", columns="ym", values="weight").fillna(0.0)
months_dt = pivot.columns.to_list()

print(f"✓ 데이터 로드 완료: {len(pivot)}개 품목, {len(months_dt)}개월")

# ============================================================================
# ⭐ 중간 품질 점수 계산 (2,700개 목표용)
# ============================================================================

def calculate_quality_score_moderate(a_series, b_series, lag):
    """
    중간 품질 점수 계산 (0-100점)
    - 원본 3,368개 → 목표 2,700개
    - 적절한 노이즈 제거 (80% 유지)
    """

    scores = {}

    # 1. 0 비율 (10점) - 중간
    zero_ratio_a = (a_series == 0).sum() / len(a_series)
    zero_ratio_b = (b_series == 0).sum() / len(b_series)
    max_zero = max(zero_ratio_a, zero_ratio_b)

    if max_zero < 0.17:
        scores['zero'] = 10
    elif max_zero < 0.37:
        scores['zero'] = 7
    elif max_zero < 0.57:  # 57%까지 허용
        scores['zero'] = 4
    else:
        scores['zero'] = 0

    # 2. 변동계수 (10점) - 중간
    cv_a = np.std(a_series) / (np.mean(a_series) + 1)
    cv_b = np.std(b_series) / (np.mean(b_series) + 1)
    max_cv = max(cv_a, cv_b)

    if max_cv < 0.75:
        scores['cv'] = 10
    elif max_cv < 1.4:
        scores['cv'] = 7
    elif max_cv < 2.7:  # 2.7까지 허용
        scores['cv'] = 4
    else:
        scores['cv'] = 0

    # 3. 스파이크 (10점) - 중간
    a_changes = np.abs(np.diff(a_series))
    b_changes = np.abs(np.diff(b_series))

    if len(a_changes) > 0 and len(b_changes) > 0:
        a_spike = a_changes.max() / (np.median(a_changes) + 1)
        b_spike = b_changes.max() / (np.median(b_changes) + 1)
        max_spike = max(a_spike, b_spike)

        if max_spike < 4.5:
            scores['spike'] = 10
        elif max_spike < 9.5:
            scores['spike'] = 7
        elif max_spike < 17:  # 17까지 허용
            scores['spike'] = 4
        else:
            scores['spike'] = 0
    else:
        scores['spike'] = 5

    # 4. 안정성 (15점) - 중간
    n = len(a_series)

    try:
        mid = n // 2

        if mid >= 12 and mid + lag < len(b_series):
            corr_first = np.corrcoef(
                a_series[:mid],
                b_series[lag:mid+lag]
            )[0, 1]

            corr_second = np.corrcoef(
                a_series[mid:],
                b_series[mid+lag:]
            )[0, 1]

            if not np.isnan(corr_first) and not np.isnan(corr_second):
                consistency = 1 - abs(corr_first - corr_second)

                if consistency > 0.52:  # 중간
                    scores['stability'] = 15
                elif consistency > 0.35:
                    scores['stability'] = 10
                elif consistency > 0.22:
                    scores['stability'] = 6
                else:
                    scores['stability'] = 3
            else:
                scores['stability'] = 5
        else:
            scores['stability'] = 5
    except:
        scores['stability'] = 5

    # 5. 트렌드 일치 (10점) - 중간
    if len(a_series) > 0 and len(b_series) > 0:
        x = np.arange(len(a_series))

        if np.std(a_series) > 0:
            slope_a, _ = np.polyfit(x, a_series, 1)
        else:
            slope_a = 0

        if np.std(b_series) > 0:
            slope_b, _ = np.polyfit(x, b_series, 1)
        else:
            slope_b = 0

        if (slope_a > 0 and slope_b > 0) or (slope_a < 0 and slope_b < 0):
            ratio = abs(slope_a) / (abs(slope_b) + 1e-10)
            if 0.38 < ratio < 2.7:  # 중간
                scores['trend'] = 10
            elif 0.25 < ratio < 4.2:
                scores['trend'] = 6
            else:
                scores['trend'] = 3
        elif slope_a == 0 or slope_b == 0:
            scores['trend'] = 5
        else:
            scores['trend'] = 2
    else:
        scores['trend'] = 4

    # 6. 방향 일치도 (10점) - 중간
    if len(a_changes) > lag and len(b_changes) > lag:
        direction_match = (np.sign(a_changes[:-lag]) == np.sign(b_changes[lag:])).mean()

        if direction_match > 0.64:  # 중간
            scores['direction'] = 10
        elif direction_match > 0.54:
            scores['direction'] = 7
        elif direction_match > 0.44:
            scores['direction'] = 4
        else:
            scores['direction'] = 1
    else:
        scores['direction'] = 4

    # 7. 통계적 유의성 (10점) - 중간
    try:
        if len(a_series) > lag:
            _, p_value = stats.pearsonr(
                a_series[:-lag] if lag > 0 else a_series,
                b_series[lag:] if lag > 0 else b_series
            )

            if p_value < 0.007:  # 중간
                scores['significance'] = 10
            elif p_value < 0.025:
                scores['significance'] = 7
            elif p_value < 0.07:
                scores['significance'] = 4
            else:
                scores['significance'] = 1
        else:
            scores['significance'] = 4
    except:
        scores['significance'] = 4

    # 8. 자기상관 (10점) - 중간
    try:
        if len(b_series) > 1:
            autocorr = np.corrcoef(b_series[:-1], b_series[1:])[0, 1]

            if 0.24 < autocorr < 0.86:  # 중간
                scores['autocorr'] = 10
            elif 0.09 < autocorr < 0.91:
                scores['autocorr'] = 7
            else:
                scores['autocorr'] = 3
        else:
            scores['autocorr'] = 4
    except:
        scores['autocorr'] = 4

    # 9. 이상치 비율 (10점) - 중간
    def count_outliers(series):
        q1 = np.percentile(series, 25)
        q3 = np.percentile(series, 75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outliers = ((series < lower) | (series > upper)).sum()
        return outliers / len(series)

    outlier_ratio_a = count_outliers(a_series)
    outlier_ratio_b = count_outliers(b_series)
    max_outlier = max(outlier_ratio_a, outlier_ratio_b)

    if max_outlier < 0.085:
        scores['outlier'] = 10
    elif max_outlier < 0.16:
        scores['outlier'] = 7
    elif max_outlier < 0.26:  # 26%까지 허용
        scores['outlier'] = 4
    else:
        scores['outlier'] = 1

    # 10. 분포 정규성 (5점) - 중간
    try:
        if len(b_series) >= 8:
            _, p_shapiro = stats.shapiro(b_series)

            if p_shapiro > 0.02:  # 중간
                scores['normality'] = 5
            elif p_shapiro > 0.004:
                scores['normality'] = 3
            else:
                scores['normality'] = 1
        else:
            scores['normality'] = 2
    except:
        scores['normality'] = 2

    total_score = sum(scores.values())

    return total_score, scores


# ============================================================================
# ⭐ 2,700개 목표 노이즈 제거
# ============================================================================

def remove_noise_pairs_for_2700(pairs, pivot, target_count=2700):
    """
    2,700개 목표 노이즈 제거

    원본: 3,368개
    목표: 2,700개 (약 80% 유지)
    전략: 적절한 노이즈 제거 (중간 모드)
    """

    print("\n" + "="*80)
    print("⭐ 2,700개 목표 노이즈 제거 (중간 모드)")
    print("="*80)
    print(f"원본: {len(pairs):,}개")
    print(f"목표: {target_count:,}개 (약 {target_count/len(pairs)*100:.0f}% 유지)")

    quality_scores = []
    detailed_scores = []

    # 품질 점수 계산 (중간 기준)
    print(f"\n품질 평가 중...")
    for row in tqdm(pairs.itertuples(), total=len(pairs), desc="평가"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)

        if leader not in pivot.index or follower not in pivot.index:
            quality_scores.append(0)
            detailed_scores.append({})
            continue

        a_series = pivot.loc[leader].values
        b_series = pivot.loc[follower].values

        total_score, scores = calculate_quality_score_moderate(a_series, b_series, lag)

        quality_scores.append(total_score)
        detailed_scores.append(scores)

    # 품질 점수 추가
    pairs_with_score = pairs.copy()
    pairs_with_score['quality_score'] = quality_scores

    # 목표 개수 기반 임계값 계산
    sorted_scores = sorted(quality_scores, reverse=True)

    if len(sorted_scores) >= target_count:
        # 목표 개수번째 점수를 임계값으로
        threshold = sorted_scores[target_count - 1]
        # 5점 단위로 내림
        threshold = max(33, (threshold // 5) * 5)  # 최소 33점
    else:
        threshold = 33  # 최소 임계값

    # 필터링
    pairs_clean = pairs_with_score[
        pairs_with_score['quality_score'] >= threshold
    ].copy()

    # 🔸 여기서는 quality_score를 남겨둬도 되고, 지금처럼 드랍해도 됨
    # (현재 요구사항은 seed 앙상블이므로 그대로 둠)
    pairs_for_model = pairs_clean.drop(columns=['quality_score'], errors='ignore')

    # 통계
    removed = len(pairs) - len(pairs_clean)
    removed_ratio = removed / len(pairs) * 100
    kept_ratio = len(pairs_clean) / len(pairs) * 100

    print(f"\n{'='*80}")
    print(f"✅ 노이즈 제거 완료!")
    print(f"{'='*80}")
    print(f"원본: {len(pairs):,}개")
    print(f"제거: {removed:,}개 ({removed_ratio:.1f}%)")
    print(f"유지: {len(pairs_clean):,}개 ({kept_ratio:.1f}%)")
    print(f"임계값: {threshold}점")

    # 점수 분포
    print(f"\n품질 점수 분포:")
    print(f"  - 평균: {np.mean(quality_scores):.1f}점")
    print(f"  - 중앙값: {np.median(quality_scores):.1f}점")
    print(f"  - 최소: {np.min(quality_scores):.1f}점")
    print(f"  - 최대: {np.max(quality_scores):.1f}점")

    # 선택된 쌍의 점수
    selected_scores = pairs_clean['quality_score'].values
    print(f"\n선택된 쌍의 점수:")
    print(f"  - 평균: {np.mean(selected_scores):.1f}점")
    print(f"  - 최소: {np.min(selected_scores):.1f}점")

    # 구간별 분포
    print(f"\n점수 구간별 분포:")
    bins = [0, 30, 40, 50, 60, 70, 80, 90, 100]
    for i in range(len(bins)-1):
        total_count = sum((np.array(quality_scores) >= bins[i]) & (np.array(quality_scores) < bins[i+1]))
        selected_count = sum((selected_scores >= bins[i]) & (selected_scores < bins[i+1]))
        print(f"  {bins[i]:3d}-{bins[i+1]:3d}점: 전체 {total_count:4,}개, 선택 {selected_count:4,}개")

    # 기준별 평균 점수
    print(f"\n기준별 평균 점수 (선택된 쌍):")
    criteria_names = {
        'zero': '0 비율',
        'cv': '변동계수',
        'spike': '스파이크',
        'stability': '안정성',
        'trend': '트렌드 일치',
        'direction': '방향 일치도',
        'significance': '통계적 유의성',
        'autocorr': '자기상관',
        'outlier': '이상치 비율',
        'normality': '정규성'
    }

    selected_indices = pairs_with_score[pairs_with_score['quality_score'] >= threshold].index
    selected_detailed = [detailed_scores[i] for i in selected_indices if i < len(detailed_scores)]

    for key, name in criteria_names.items():
        scores = [s.get(key, 0) for s in selected_detailed if s]
        if scores:
            if key == 'stability':
                max_score = 15
            elif key == 'normality':
                max_score = 5
            else:
                max_score = 10

            print(f"  {name:15s}: 평균 {np.mean(scores):5.1f}점 (최대 {max_score}점)")

    return pairs_for_model


# ============================================================================
# 조건부 Weight 피처
# ============================================================================

def create_conditional_weight_features(value, weight, value_series):
    """조건부 Weight 피처 4개"""
    value_median = np.median(value_series[value_series > 0]) if (value_series > 0).any() else 0
    value_q75 = np.percentile(value_series[value_series > 0], 75) if (value_series > 0).any() else 0

    if value > value_median:
        weight_for_high_value = weight * 1.5
    else:
        weight_for_high_value = 0

    weight_value_ratio = weight * (value / (value_median + 1))
    exp_weighted = weight * (1 - np.exp(-value / (value_median + 1)))

    if value > value_q75:
        weight_importance = weight * 1.5
    elif value > value_median:
        weight_importance = weight * 1.0
    else:
        weight_importance = weight * 0.1

    return {
        'weight_for_high_value': weight_for_high_value,
        'weight_value_ratio': weight_value_ratio,
        'exp_weighted': exp_weighted,
        'weight_importance': weight_importance
    }


# ============================================================================
# 학습 데이터 구축 (32개 피처)
# ============================================================================

def build_training_data_optimized(pivot, pivot_weight, pairs, months_dt):
    """32개 피처 학습 데이터 생성"""
    months = months_dt
    n_months = len(months)
    rows = []

    for row in pairs.itertuples(index=False):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        for t in range(max(lag, 12), n_months - 1):
            b_t = b_series[t]
            b_t_1 = b_series[t - 1]
            b_t_2 = b_series[t - 2]
            b_t_12 = b_series[t - 12]
            a_t_lag = a_series[t - lag]
            b_t_plus_1 = b_series[t + 1]
            b_weight_t = b_weight_series[t]
            target_month = months[t + 1].month

            a_t_lag_diff = a_series[t - lag] - a_series[t - lag - 1]
            a_momentum = a_series[t - lag] / (a_series[t - lag - 3] + 1)
            a_roll_mean_3 = (a_series[t-lag] + a_series[t-lag-1] + a_series[t-lag-2]) / 3

            b_diff = b_t - b_t_1
            b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
            b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
            b_roll_mean_3 = (b_series[t] + b_series[t-1] + b_series[t-2]) / 3
            b_roll_mean_6 = (b_series[t-5:t+1]).mean()
            b_roll_std_3 = np.std([b_series[t], b_series[t-1], b_series[t-2]])

            ab_ratio = a_t_lag / (b_t + 1)
            ab_diff = a_t_lag - b_t
            corr_weighted_a = a_t_lag * corr

            b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
            b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
            b_yoy_ratio = b_t / (b_t_12 + 1)
            b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
            recent_12 = b_series[max(0, t-11):t+1]
            b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

            quarter = (target_month - 1) // 3 + 1
            is_year_end = 1 if target_month in [11, 12, 1] else 0
            lag_weight = 1 / (1 + lag)

            b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

            rows.append({
                "b_t": b_t, "b_t_1": b_t_1, "b_t_12": b_t_12, "a_t_lag": a_t_lag,
                "max_corr": corr, "best_lag": float(lag), "month": float(target_month),
                "a_t_lag_diff": a_t_lag_diff, "a_momentum": a_momentum, "a_roll_mean_3": a_roll_mean_3,
                "b_diff": b_diff, "b_pct_change": b_pct_change, "b_yoy_growth": b_yoy_growth,
                "b_roll_mean_3": b_roll_mean_3, "b_roll_mean_6": b_roll_mean_6, "b_roll_std_3": b_roll_std_3,
                "ab_ratio": ab_ratio, "ab_diff": ab_diff, "corr_weighted_a": corr_weighted_a,
                "b_cv": b_cv, "b_acceleration": b_acceleration, "b_yoy_ratio": b_yoy_ratio,
                "b_ma_ratio": b_ma_ratio, "b_percentile_rank": b_percentile_rank,
                "quarter": float(quarter), "is_year_end": is_year_end, "lag_weight": lag_weight,
                "b_weight_for_high_value": b_weight_features['weight_for_high_value'],
                "b_weight_value_ratio": b_weight_features['weight_value_ratio'],
                "b_exp_weighted": b_weight_features['exp_weighted'],
                "b_weight_importance": b_weight_features['weight_importance'],
                "target": np.log1p(b_t_plus_1),
            })

    return pd.DataFrame(rows)


# ============================================================================
# 예측 함수 (seed 앙상블 버전)
# ============================================================================

def predict_optimized(pivot, pivot_weight, pairs, model_lgb_list, model_xgb_list, months_dt):
    """32개 피처 예측 + LGBM/XGB seed 앙상블 + 0.7/0.3 앙상블"""
    months = months_dt
    n_months = len(months)
    t_last = n_months - 1
    t_prev = n_months - 2
    preds = []
    target_month = months[-1].month + 1 if months[-1].month < 12 else 1

    for row in tqdm(pairs.itertuples(index=False), desc="예측"):
        leader = row.leading_item_id
        follower = row.following_item_id
        lag = int(row.best_lag)
        corr = float(row.max_corr)

        if leader not in pivot.index or follower not in pivot.index:
            continue
        if leader not in pivot_weight.index or follower not in pivot_weight.index:
            continue

        a_series = pivot.loc[leader].values.astype(float)
        b_series = pivot.loc[follower].values.astype(float)
        b_weight_series = pivot_weight.loc[follower].values.astype(float)

        if t_last - lag < 0 or t_last < 12:
            continue

        b_t = b_series[t_last]
        b_t_1 = b_series[t_prev]
        b_t_2 = b_series[t_prev - 1]
        b_t_12 = b_series[t_last - 12]
        a_t_lag = a_series[t_last - lag]
        b_weight_t = b_weight_series[t_last]

        a_t_lag_diff = a_series[t_last - lag] - a_series[t_last - lag - 1]
        a_momentum = a_series[t_last - lag] / (a_series[t_last - lag - 3] + 1)
        a_roll_mean_3 = (a_series[t_last-lag] + a_series[t_last-lag-1] + a_series[t_last-lag-2]) / 3

        b_diff = b_t - b_t_1
        b_pct_change = (b_t - b_t_1) / (b_t_1 + 1)
        b_yoy_growth = (b_t - b_t_12) / (b_t_12 + 1)
        b_roll_mean_3 = (b_series[t_last] + b_series[t_last-1] + b_series[t_last-2]) / 3
        b_roll_mean_6 = b_series[t_last-5:t_last+1].mean()
        b_roll_std_3 = np.std([b_series[t_last], b_series[t_last-1], b_series[t_last-2]])

        ab_ratio = a_t_lag / (b_t + 1)
        ab_diff = a_t_lag - b_t
        corr_weighted_a = a_t_lag * corr

        b_cv = b_roll_std_3 / (b_roll_mean_3 + 1)
        b_acceleration = (b_t - b_t_1) - (b_t_1 - b_t_2)
        b_yoy_ratio = b_t / (b_t_12 + 1)
        b_ma_ratio = b_roll_mean_3 / (b_roll_mean_6 + 1)
        recent_12 = b_series[max(0, t_last-11):t_last+1]
        b_percentile_rank = (b_t >= recent_12).sum() / len(recent_12)

        quarter = (target_month - 1) // 3 + 1
        is_year_end = 1 if target_month in [11, 12, 1] else 0
        lag_weight = 1 / (1 + lag)

        b_weight_features = create_conditional_weight_features(b_t, b_weight_t, b_series)

        X_test = np.array([[
            b_t, b_t_1, b_t_12, a_t_lag, corr, float(lag), float(target_month),
            a_t_lag_diff, a_momentum, a_roll_mean_3,
            b_diff, b_pct_change, b_yoy_growth, b_roll_mean_3, b_roll_mean_6, b_roll_std_3,
            ab_ratio, ab_diff, corr_weighted_a,
            b_cv, b_acceleration, b_yoy_ratio, b_ma_ratio, b_percentile_rank,
            float(quarter), is_year_end, lag_weight,
            b_weight_features['weight_for_high_value'],
            b_weight_features['weight_value_ratio'],
            b_weight_features['exp_weighted'],
            b_weight_features['weight_importance']
        ]])

        # 🔹 여러 seed 모델의 예측을 평균
        pred_lgb_list = [m.predict(X_test)[0] for m in model_lgb_list]
        pred_xgb_list = [m.predict(X_test)[0] for m in model_xgb_list]

        pred_lgb = float(np.mean(pred_lgb_list))
        pred_xgb = float(np.mean(pred_xgb_list))

        # 기존과 동일한 0.7 / 0.3 앙상블
        y_pred_log = 0.7 * pred_lgb + 0.3 * pred_xgb
        y_pred = np.expm1(y_pred_log)
        y_pred = max(0.0, float(y_pred))
        y_pred = int(round(y_pred))

        preds.append({
            "leading_item_id": leader,
            "following_item_id": follower,
            "value": y_pred,
        })

    return pd.DataFrame(preds)


# ============================================================================
# 메인 실행
# ============================================================================

if __name__ == "__main__":

    print("\n" + "="*80)
    print("⭐ 2,700개 목표 최적화 버전 (seed 앙상블 포함)")
    print("="*80)

    # ⭐ 2,700개 목표 노이즈 제거
    pairs_for_model = remove_noise_pairs_for_2700(
        pairs=pairs_with_dtw,
        pivot=pivot,
        target_count=2700      # ⭐ 목표: 2,700개
    )

    print(f"\n최종 사용 쌍: {len(pairs_for_model)}개")

    # 학습
    print("\n" + "="*80)
    print("학습 데이터 생성")
    print("="*80)

    df_train_model = build_training_data_optimized(pivot, pivot_weight, pairs_for_model, months_dt)
    print(f"✓ 학습 데이터: {df_train_model.shape}")

    feature_cols = [
        'b_t', 'b_t_1', 'b_t_12', 'a_t_lag', 'max_corr', 'best_lag', 'month',
        'a_t_lag_diff', 'a_momentum', 'a_roll_mean_3',
        'b_diff', 'b_pct_change', 'b_yoy_growth', 'b_roll_mean_3', 'b_roll_mean_6', 'b_roll_std_3',
        'ab_ratio', 'ab_diff', 'corr_weighted_a',
        'b_cv', 'b_acceleration', 'b_yoy_ratio', 'b_ma_ratio', 'b_percentile_rank',
        'quarter', 'is_year_end', 'lag_weight',
        'b_weight_for_high_value', 'b_weight_value_ratio', 'b_exp_weighted', 'b_weight_importance'
    ]

    if not df_train_model.empty:
        print("\n" + "="*80)
        print("모델 학습 (여러 seed 앙상블)")
        print("="*80)

        train_X = df_train_model[feature_cols].values
        train_y = df_train_model["target"].values

        print(f"학습 샘플: {len(train_X):,}, 피처: {len(feature_cols)}")

        # 🔹 여러 seed로 LGBM/XGB 학습
        seeds_lgb = [42, 2024, 999]
        seeds_xgb = [42, 2024, 999]

        lgb_models = []
        xgb_models = []

        print("\nLGBM seed 앙상블 학습 중...")
        for s in seeds_lgb:
            print(f"  - LGBM (seed={s}) 학습 중...")
            model_lgb = LGBMRegressor(
                random_state=s,
                n_estimators=700, learning_rate=0.01, max_depth=8,
                num_leaves=50, min_child_samples=20,
                subsample=0.8, colsample_bytree=0.8,
                reg_alpha=0.1, reg_lambda=0.1, min_split_gain=0.01
            )
            model_lgb.fit(train_X, train_y)
            lgb_models.append(model_lgb)
        print("✓ LGBM seed 앙상블 완료")

        print("\nXGBoost seed 앙상블 학습 중...")
        for s in seeds_xgb:
            print(f"  - XGB (seed={s}) 학습 중...")
            model_xgb = XGBRegressor(
                random_state=s,
                n_estimators=900, learning_rate=0.02, max_depth=8,
                subsample=0.8, colsample_bytree=0.8,
                reg_alpha=0.1, reg_lambda=0.1,
                tree_method='hist'
            )
            model_xgb.fit(train_X, train_y)
            xgb_models.append(model_xgb)
        print("✓ XGBoost seed 앙상블 완료")

        print("\n" + "="*80)
        print("예측")
        print("="*80)

        submission = predict_optimized(pivot, pivot_weight, pairs_for_model, lgb_models, xgb_models, months_dt)
        print(f"\n✓ 예측 완료: {len(submission)}개")

        date_str = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

        submission.to_csv(f'_result_sy_2700_moderate_seedEns_{date_str}.csv', index=False)

        print("\n✅ 2,700개 목표 버전 (seed 앙상블) 완료!")
        print("  - 원본: 3,368개")
        print("  - 목표: 2,700개")
        print("  - 중간 기준 (적절한 노이즈 제거)")
        print("  - 32개 피처")
        print("  - LGBM + XGB + seed 앙상블 (0.7 / 0.3)")
        print("="*80)
```
